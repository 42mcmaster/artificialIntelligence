## Experiment 1: Learning the Circle Dataset

**Setup:**
- Select **circle dataset** (bottom left under "DATA")
- Set hidden layers to **1 layer with 2 neurons**
- Click **Play** and let it train for about 20 seconds

**Observation Questions:**

1. Did the network successfully learn to separate the blue and orange points?
   **Answer:** No, 2 neurons cannot learn a circle. You'll see a single curved line or linear boundary.

2. What did the learned boundary look like? (Straight line? Curved? Multiple regions?)
   **Answer:** Mostly a diagonal or slightly curved line. Cannot create the circular decision boundary needed.

3. Why do you think 2 neurons wasn't enough? Explain in 1-2 sentences.
   **Answer:** A single layer with 2 neurons is too simple. It can only create linear or slightly curved boundaries. A circle is a non-linear pattern that requires more complexity to represent.

### Instructor Note:
This teaches the concept of model capacity. Underfitting occurs when the model is too simple for the problem.

## Experiment 2: Increasing Neuron Count

**Setup:**
- Keep the **circle dataset**
- Change to **1 layer with 8 neurons**
- Click **Play** and let it train for about 20 seconds

**Observation Questions:**

1. How is the learned boundary different from Experiment 1?
   **Answer:** Much more complex! The boundary now curves and bends to approximate the circular shape. Multiple neurons can create more complex, non-linear boundaries.

2. Did the network learn faster (fewer iterations)? Slower?
   **Answer:** Typically learns faster - more neurons = more flexibility = can find solution quicker. Usually converges in 200-400 iterations vs 1000+ for 2 neurons.

3. Can 8 neurons solve the circle problem? Why or why not?
   **Answer:** Yes, mostly! A single layer with 8 neurons can create complex boundaries. It may not be perfect, but significantly better than 2 neurons.

### Instructor Note:
More neurons = greater model capacity. But with one layer, there's still a limit to complexity.

## Experiment 3: Adding a Hidden Layer

**Setup:**
- Keep the **circle dataset**
- Change to **2 hidden layers with 8 neurons each**
  - (Click the **+** button to add a second hidden layer)
- Click **Play** and let it train

**Observation Questions:**

1. Does having 2 layers with 8 neurons each learn better than 1 layer with 8 neurons?
   **Answer:** Yes! Multiple layers can compose simpler patterns into more complex ones. May learn slightly faster.

2. How many total neurons are we using now? Calculate it:
   - Layer 1: 8 neurons
   - Layer 2: 8 neurons
   - Total: 16 neurons

3. Why might multiple layers help even if they have fewer total neurons?
   **Answer:** Depth matters! A 3-layer network with 5 neurons each (15 total) can learn more complex patterns than a 1-layer network with 16 neurons. Layers can learn hierarchical features.

### Instructor Note:
This demonstrates why depth matters. Multiple layers create hierarchical representations that are more efficient than a single wide layer.

## Experiment 5: Deep Network for Spiral

**Setup:**
- Keep the **spiral dataset**
- Change to **3 hidden layers with 16 neurons each**
  - (Click **+** twice to add two more layers)
- Click **Play** and let it train

**Observation Questions:**

1. Does the deep network solve the spiral better? How much better?
   **Answer:** YES, dramatically better! A 3-layer network can achieve near-perfect classification of the spiral in 1000-2000 iterations. This is the power of depth.

2. How many total neurons are you using now?
   - Calculation: 3 layers × 16 neurons = 48 total

3. Why does the spiral need a deeper (more layers) network?
   **Answer:** The spiral has an extremely non-linear, hierarchical structure. Layer 1 learns simple curved boundaries. Layer 2 combines those into more complex shapes. Layer 3 combines those into the final intertwined spiral separation. Shallow networks can't compose these abstractions.

### Instructor Note:
This is THE key lesson about deep learning. More layers enable learning hierarchical, abstract representations that shallow networks cannot capture.

## Summary Analysis

Based on all your experiments, answer these:

1. **What did you learn about the relationship between neurons and learning?**
   **Answer:** More neurons in a layer increase the capacity to learn more complex, non-linear boundaries. However, there are diminishing returns. 4 neurons might be twice as good as 2, but 8 isn't 4x better than 4.

2. **What did you learn about the relationship between layers (depth) and learning?**
   **Answer:** More layers enable learning of hierarchical, abstract patterns. A 2-layer network can solve problems a 1-layer network cannot, even with fewer total neurons. Very complex patterns (spiral) need 3+ layers.

3. **When would you use a deep network (many layers) vs a wide network (many neurons per layer)?**
   **Answer:** Use deep networks for hierarchical, very non-linear data (images, text, spiral-like patterns). Use wide networks for simpler problems or when you have limited data (to avoid overfitting). There's a tradeoff: depth learns better patterns, width learns simpler ones faster.

4. **Is there a point where adding more neurons/layers doesn't help? Why?**
   **Answer:** YES! Eventually you hit:
   - Overfitting: Too much capacity learns noise instead of patterns
   - Computational limits: Training gets very slow
   - The actual complexity ceiling: Some problems have inherent limits
   The key is finding the right balance.

### Instructor Note:
These answers teach essential deep learning principles without getting into formal complexity theory. The Playground provides visual intuition for concepts students will encounter in CS/ML careers.
