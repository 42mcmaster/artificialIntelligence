## Purpose

Practice writing SQL aggregation queries using GROUP BY, COUNT, SUM, AVG, MIN, MAX, and HAVING.

## Key Concepts

**Aggregate Functions:**
- `COUNT(*)` — Count number of rows
- `SUM()` — Add up values
- `AVG()` — Calculate average
- `MIN()` — Find minimum value
- `MAX()` — Find maximum value

**GROUP BY** — Summarize data by category

**HAVING** — Filter groups (like WHERE but for groups)

**Example:**
```sql
SELECT position, COUNT(*) as playerCount, AVG(pts) as avgPoints
FROM player_season_stats
WHERE season = '2021-22'
GROUP BY position
HAVING COUNT(*) > 5
ORDER BY avgPoints DESC
```
