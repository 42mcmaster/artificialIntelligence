## Setup

````python
import pandas as pd                    # pandas for DataFrames
import sqlite3                         # sqlite3 for database connection

nbaConnection = sqlite3.connect('nba_5seasons.db')   # Connect to database
print("✅ Connected to NBA database")
````

## Question 1: Find All Florida Teams

````python
# Query to find all Florida teams
floridaTeamsQuery = """
SELECT full_name, city, state
FROM teams
WHERE state = 'Florida'
"""

floridaTeams = pd.read_sql(floridaTeamsQuery, nbaConnection)
display(floridaTeams)
print(f"\nFound {len(floridaTeams)} Florida NBA teams")
````

### Instructor Note
- Simple WHERE with text filter
- Returns 2 teams: Heat (Miami) and Magic (Orlando)
- Note the variable name: floridaTeams (camelCase)

## Question 2: Teams Founded in 1995

````python
# Query to find teams founded in 1995
foundedQuery = """
SELECT full_name, city, year_founded
FROM teams
WHERE year_founded = 1995
"""

foundedTeams = pd.read_sql(foundedQuery, nbaConnection)
display(foundedTeams)
print(f"\nFound {len(foundedTeams)} teams founded in 1995")
````

### Instructor Note
- WHERE with numeric comparison (no quotes needed)
- Returns 2 teams: Raptors and Grizzlies
- Both Canadian and Memphis teams joined in 1995

## Question 3: Top 5 Oldest Teams

````python
# Query to find 5 oldest NBA teams
oldestTeamsQuery = """
SELECT full_name, year_founded
FROM teams
ORDER BY year_founded
LIMIT 5
"""

oldestTeams = pd.read_sql(oldestTeamsQuery, nbaConnection)
display(oldestTeams)
print(f"\nOldest NBA team: {oldestTeams.iloc[0]['full_name']} ({oldestTeams.iloc[0]['year_founded']})")
````

### Instructor Note
- ORDER BY without DESC = ascending (oldest first)
- LIMIT 5 returns only first 5
- Three teams tied for oldest (1946): Celtics, Warriors, Knicks
- Important pattern: ORDER BY comes BEFORE LIMIT

## Question 4: Teams from New York or California

````python
# Query to find teams from NY or California
coastTeamsQuery = """
SELECT full_name, city, state
FROM teams
WHERE state = 'California' OR state = 'New York'
ORDER BY state, city
"""

coastTeams = pd.read_sql(coastTeamsQuery, nbaConnection)
display(coastTeams)
print(f"\nFound {len(coastTeams)} teams in California or New York")
````

### Instructor Note
- OR combines two conditions (at least one must be true)
- Returns 6 teams: 4 from CA (Warriors, Lakers, Clippers, Kings), 2 from NY (Knicks, Nets)
- Added ORDER BY state, city for cleaner display
- Shows multi-column sorting (primary then secondary)

## Question 5: High-Scoring Players in 2021-22

````python
# Query to find 25+ PPG scorers in 2021-22 season
topScorersQuery = """
SELECT player_id, season, pts, gp
FROM player_season_stats
WHERE season = '2021-22' AND pts >= 25 AND gp >= 40
ORDER BY pts DESC
LIMIT 10
"""

topScorers = pd.read_sql(topScorersQuery, nbaConnection)
display(topScorers)
print(f"\nFound {len(topScorers)} players averaging 25+ PPG in 2021-22")
print(f"Top scorer averaged {topScorers.iloc[0]['pts']} PPG")
````

### Instructor Note
- Multiple AND conditions: season, points, games played
- gp >= 40 ensures meaningful playing time
- ORDER BY pts DESC gets highest scorers first
- LIMIT 10 returns top 10
- Real ML-ready query pattern

## Question 6: High-Scoring Games in 2021-22

````python
# Query to find top 15 high-scoring games in 2021-22
highScoringQuery = """
SELECT game_date, team_id, pts, wl
FROM team_game_stats
WHERE season = '2021-22' AND pts >= 130
ORDER BY pts DESC
LIMIT 15
"""

highScoringGames = pd.read_sql(highScoringQuery, nbaConnection)
display(highScoringGames)
print(f"\nFound {len(highScoringGames)} games with 130+ points in 2021-22")
print(f"Highest score: {highScoringGames.iloc[0]['pts']} points")
````

### Instructor Note
- Uses team_game_stats table (one row per team per game)
- WHERE filters for season and point threshold
- ORDER BY DESC sorts highest to lowest
- LIMIT 15 returns top 15 games
- Shows mix of wins and losses (wl column)

## Question 7: Elite Shooters

````python
# Query to find elite 3-point shooters in 2021-22
eliteShootersQuery = """
SELECT player_id, fg3_pct, pts, gp
FROM player_season_stats
WHERE season = '2021-22' AND fg3_pct >= 40 AND gp >= 40 AND pts >= 10
ORDER BY fg3_pct DESC
LIMIT 10
"""

eliteShooters = pd.read_sql(eliteShootersQuery, nbaConnection)
display(eliteShooters)
print(f"\nFound {len(eliteShooters)} elite 3-point shooters")
if len(eliteShooters) > 0:
    print(f"Best 3PT%: {eliteShooters.iloc[0]['fg3_pct']:.1f}%")
````

### Instructor Note
- Complex query with 4 AND conditions
- fg3_pct = 3-point shooting percentage
- gp >= 40 ensures meaningful sample
- pts >= 10 ensures role player threshold
- ORDER BY fg3_pct DESC shows best shooters first
- Shows how to build complex WHERE clauses

## Question 8: Teams Founded Between 1970 and 1990

````python
# Query to find teams founded 1970-1990 using BETWEEN
eraTeamsQuery = """
SELECT full_name, year_founded, state
FROM teams
WHERE year_founded BETWEEN 1970 AND 1990
ORDER BY year_founded
"""

eraTeams = pd.read_sql(eraTeamsQuery, nbaConnection)
display(eraTeams)
print(f"\nFound {len(eraTeams)} teams founded between 1970-1990")
````

### Instructor Note
- BETWEEN x AND y is shorthand for: >= x AND <= y
- More readable than: WHERE year_founded >= 1970 AND year_founded <= 1990
- Includes both endpoints (1970 and 1990)
- ORDER BY year_founded shows chronological order
- Returns 11 teams from this era

## Summary of Key Concepts

### Question 1-2: Basic WHERE Filtering
- Text needs single quotes: `'Florida'`
- Numbers don't: `1995`

### Question 3: ORDER BY + LIMIT Pattern
- ORDER BY comes before LIMIT
- Default is ascending (oldest first)
- Perfect for "Top N" queries

### Question 4: OR Logic
- OR requires at least one condition true
- Different from AND (which requires both)
- Optional: Add ORDER BY to organize results

### Question 5-7: Complex WHERE with Multiple AND
- Can chain multiple AND conditions
- Each condition must be true for row to be included
- Pattern: WHERE col1 = x AND col2 > y AND col3 >= z
- Always include gp >= 40 for player stats (meaningful playing time)

### Question 8: BETWEEN for Ranges
- Cleaner than: col >= min AND col <= max
- Inclusive of both endpoints
- Works for dates, numbers, any comparable values

## All Queries Share Pattern

```
SELECT (which columns)
FROM (which table)
WHERE (filter rows)
ORDER BY (sort results)
LIMIT (how many)
```

This is the foundation for all SQL queries in this course!
