# PART 3: WHERE Clause — Filtering Rows

## Filtering with Text Values

````python
# Find teams from a specific state (text needs single quotes!)
texasQuery = """
SELECT full_name, city, state
FROM teams
WHERE state = 'Texas'
"""

texasTeams = pd.read_sql(texasQuery, nbaConnection)   # Execute query
print("Texas NBA Teams:")
display(texasTeams)
print(f"\nTexas has {len(texasTeams)} NBA teams")
````

### Instructor Note
- WHERE clause filters rows before returning them
- Text values MUST use single quotes: 'Texas', NOT Texas
- Using = (not ==) which is SQL syntax, not Python syntax
- Returns only the 3 teams in Texas: Mavericks, Rockets, Spurs

## Filtering with Numbers

````python
# Find teams founded after a certain year (numbers don't need quotes!)
newerQuery = """
SELECT full_name, year_founded
FROM teams
WHERE year_founded > 1980
"""

newerTeams = pd.read_sql(newerQuery, nbaConnection)   # Execute query
print("Teams Founded After 1980:")
display(newerTeams)
````

### Instructor Note
- Numbers do NOT need single quotes
- Using > (greater than) operator
- Returns 9 teams founded after 1980
- Can also use <, >=, <=, !=, =

## Try This: Filter Teams

````python
# Solution: Find all teams from California
californiaQuery = """
SELECT full_name, city
FROM teams
WHERE state = 'California'
"""

californiaTeams = pd.read_sql(californiaQuery, nbaConnection)   # Execute query
display(californiaTeams)
````

# PART 4: Combining Conditions with AND & OR

## AND — Both Conditions Must Be True

````python
# Find old California teams (both conditions must be true)
oldCalQuery = """
SELECT full_name, city, state, year_founded
FROM teams
WHERE state = 'California' AND year_founded < 1970
"""

oldCalTeams = pd.read_sql(oldCalQuery, nbaConnection)   # Execute query
print("Old California Teams (founded before 1970):")
display(oldCalTeams)
````

### Instructor Note
- AND requires BOTH conditions to be true
- Returns only 3 teams: Warriors (1946), Lakers (1948), Kings (1948)
- More restrictive than OR
- Can chain multiple AND conditions

## OR — At Least One Condition Must Be True

````python
# Find teams from California or Texas (at least one condition true)
caTexasQuery = """
SELECT full_name, city, state
FROM teams
WHERE state = 'California' OR state = 'Texas'
"""

caTexasTeams = pd.read_sql(caTexasQuery, nbaConnection)   # Execute query
print("Teams from California OR Texas:")
display(caTexasTeams)
````

### Instructor Note
- OR requires at least ONE condition to be true
- Returns 7 teams total (4 from CA + 3 from TX)
- Less restrictive than AND
- Use when you want ANY matching criteria

## Try This: Multiple Conditions

````python
# Solution: Find teams founded between 1970 and 1990
betweenQuery = """
SELECT full_name, year_founded
FROM teams
WHERE year_founded > 1970 AND year_founded < 1990
"""

betweenTeams = pd.read_sql(betweenQuery, nbaConnection)   # Execute query
display(betweenTeams)
````

# PART 5: ORDER BY — Sorting Results

## Sort Ascending (Default)

````python
# Sort teams by founding year (oldest first - ascending is default)
oldestQuery = """
SELECT full_name, year_founded
FROM teams
ORDER BY year_founded
"""

oldestTeams = pd.read_sql(oldestQuery, nbaConnection)   # Execute query
print("Teams Sorted by Age (Oldest First):")
display(oldestTeams)
````

### Instructor Note
- Default sort order is ascending (smallest to largest)
- No need to write ASC (it's assumed)
- Shows Celtics, Warriors, Knicks all from 1946

## Sort Descending (DESC)

````python
# Sort teams by founding year (newest first - using DESC)
newestQuery = """
SELECT full_name, year_founded
FROM teams
ORDER BY year_founded DESC
"""

newestTeams = pd.read_sql(newestQuery, nbaConnection)   # Execute query
print("Teams Sorted by Age (Newest First):")
display(newestTeams)
````

### Instructor Note
- DESC must be explicitly written (ascending is default)
- DESC = descending (largest to smallest)
- Pelicans (2002) are the newest franchise

## Sort by Multiple Columns

````python
# Sort by primary column (state), then secondary column (city)
multiSortQuery = """
SELECT full_name, city, state
FROM teams
ORDER BY state, city
"""

multiSort = pd.read_sql(multiSortQuery, nbaConnection)   # Execute query
print("Teams Sorted by State, then City:")
display(multiSort)
````

### Instructor Note
- Sorts by first column (state) alphabetically
- Within each state, sorts by second column (city)
- Arizona first, then California teams sorted by city
- Powerful for organizing hierarchical data

## Try This: Sort Data

````python
# Solution: Sort teams by city name (alphabetically)
citySortQuery = """
SELECT full_name, city
FROM teams
ORDER BY city
"""

citySort = pd.read_sql(citySortQuery, nbaConnection)   # Execute query
display(citySort)
````

# PART 6: LIMIT — Controlling Output Size

## Get First N Rows

````python
# Get first 5 teams
limitQuery = """
SELECT *
FROM teams
LIMIT 5
"""

firstFive = pd.read_sql(limitQuery, nbaConnection)   # Execute query
print("First 5 Teams:")
display(firstFive)
````

### Instructor Note
- LIMIT restricts the number of rows returned
- LIMIT 5 means "return only the first 5 rows"
- Useful for testing queries on large datasets
- Always goes at the very end of the query

## Top N Query

````python
# Get oldest 5 teams (ORDER BY first, then LIMIT)
topOldestQuery = """
SELECT full_name, year_founded
FROM teams
ORDER BY year_founded
LIMIT 5
"""

topOldest = pd.read_sql(topOldestQuery, nbaConnection)   # Execute query
print("Top 5 Oldest NBA Teams:")
display(topOldest)
````

### Instructor Note
- ORDER BY is applied BEFORE LIMIT
- This is how you get "Top N" results
- Common pattern: ORDER BY DESC LIMIT 10 to get top 10 by value

## Try This: Top N

````python
# Solution: Get the 10 newest teams
topNewestQuery = """
SELECT full_name, year_founded
FROM teams
ORDER BY year_founded DESC
LIMIT 10
"""

topNewest = pd.read_sql(topNewestQuery, nbaConnection)   # Execute query
display(topNewest)
````

# PART 8: Complete Query Examples

## Example 1: Top Scorers in 2021-22 Season

````python
# Find top 10 scorers who played substantial minutes
topScorersQuery = """
SELECT player_id, pts, reb, ast, gp
FROM player_season_stats
WHERE season = '2021-22' AND pts >= 15 AND gp >= 40
ORDER BY pts DESC
LIMIT 10
"""

topScorers = pd.read_sql(topScorersQuery, nbaConnection)   # Execute query
print("Top 10 Scorers in 2021-22 Season:")
display(topScorers)
````

### Instructor Note
- gp >= 40 ensures meaningful playing time (out of ~82 games)
- WHERE conditions are applied before ORDER BY and LIMIT
- This is a real ML-ready query that filters, sorts, and limits
- Students should understand each clause's role

## Example 2: Explore Game Statistics

````python
# See what columns are in team_game_stats
gameStructureQuery = "PRAGMA table_info(team_game_stats)"

gameStructure = pd.read_sql(gameStructureQuery, nbaConnection)   # Execute query
print("Team Game Stats Table Structure:")
display(gameStructure)
````

### Instructor Note
- One row per team per game (not per player)
- Key columns: season, team_id, game_date, wl (win/loss), pts
- Perfect for training win prediction models

## Example 3: High-Scoring Games

````python
# Find highest-scoring team performances in 2021-22
highScoringQuery = """
SELECT game_date, team_id, pts, wl
FROM team_game_stats
WHERE season = '2021-22' AND pts >= 125
ORDER BY pts DESC
LIMIT 15
"""

highScoringGames = pd.read_sql(highScoringQuery, nbaConnection)   # Execute query
print("Highest-Scoring Team Performances (2021-22):")
display(highScoringGames)
````

### Instructor Note
- Shows real games from 2021-22 season
- team_id maps to teams table for full team names
- wl shows whether they won or lost despite high scoring
- Good example of filtering for interesting data

## Example 4: Wins vs Losses

````python
# Find games won with good offensive scoring
winsQuery = """
SELECT game_date, team_id, pts, wl
FROM team_game_stats
WHERE season = '2021-22' AND wl = 'W'
ORDER BY pts DESC
LIMIT 20
"""

winningGames = pd.read_sql(winsQuery, nbaConnection)   # Execute query
print("Winning Games in 2021-22 Season (Sorted by Points):")
display(winningGames)
````

### Instructor Note
- WHERE wl = 'W' filters to wins only
- wl is text ('W' or 'L') so needs quotes
- Shows that high-scoring usually correlates with winning
- Good for building intuition about what features matter for ML

# PART 9: BETWEEN — Range Filtering

## BETWEEN for Ranges

````python
# Find games with moderate scoring (100-110 points)
moderateQuery = """
SELECT game_date, team_id, pts, wl
FROM team_game_stats
WHERE season = '2021-22' AND pts BETWEEN 100 AND 110
ORDER BY pts DESC
LIMIT 20
"""

moderateGames = pd.read_sql(moderateQuery, nbaConnection)   # Execute query
print("Games with 100-110 Points (2021-22):")
display(moderateGames)
````

### Instructor Note
- BETWEEN x AND y is shorthand for: >= x AND <= y
- Includes both endpoints (100 and 110)
- More readable than multiple AND conditions
- Useful for age ranges, score ranges, date ranges, etc.

# PART 11: Summary & Query Template

## Complete SQL Query Template

```sql
SELECT   column1, column2, column3    -- WHAT columns to show
FROM     table_name                   -- WHERE to get data
WHERE    condition                    -- FILTER rows
ORDER BY column_name                  -- SORT results
LIMIT    10                           -- HOW MANY rows
```

**Order matters!** Always use this sequence.

## Python Pattern

````python
# STEP 1: Write SQL query as triple-quoted string
sampleQuery = """
SELECT column1, column2
FROM table_name
WHERE condition
ORDER BY column
LIMIT 10
"""

# STEP 2: Execute query with pd.read_sql()
resultDataFrame = pd.read_sql(sampleQuery, nbaConnection)   # Execute and load into DataFrame

# STEP 3: Display results
display(resultDataFrame)   # Show the data

# STEP 4: Analyze or use for ML
print(f"Found {len(resultDataFrame)} rows")   # Quick analysis

# STEP 5: Close connection when completely done
# nbaConnection.close()   # Uncomment at the very end
````

### Instructor Note
- Triple quotes (""" """) allow multi-line strings
- camelCase for variables (sampleQuery, resultDataFrame)
- pd.read_sql() returns a proper pandas DataFrame
- Don't close connection until all queries are complete
