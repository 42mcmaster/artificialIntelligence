# PART 11: Summary & Query Template

## Complete SQL Query Template

```sql
SELECT   column1, column2, column3    -- What columns to show
FROM     table_name                   -- Which table to use
WHERE    condition                    -- Filter rows (optional)
ORDER BY column_name                  -- Sort results (optional)
LIMIT    10                           -- Limit results (optional)
```

### Order MATTERS!
1. SELECT (what columns)
2. FROM (which table)
3. WHERE (filter rows)
4. ORDER BY (sort)
5. LIMIT (limit count)

## Comparison Operators

| Operator | Meaning | Example |
|----------|---------|----------|
| `=` | Equal | `WHERE state = 'Ohio'` |
| `!=` | Not equal | `WHERE state != 'California'` |
| `>` | Greater than | `WHERE pts > 20` |
| `<` | Less than | `WHERE year_founded < 1980` |
| `>=` | Greater or equal | `WHERE gp >= 40` |
| `<=` | Less or equal | `WHERE age <= 30` |
| `BETWEEN` | In range | `WHERE pts BETWEEN 15 AND 25` |

---

## Common Mistakes

### ❌ Forgetting quotes around text
```sql
WHERE state = California  -- WRONG!
WHERE state = 'California'  -- RIGHT
```

### ❌ Wrong clause order
```sql
WHERE state = 'Texas' FROM teams SELECT full_name  -- WRONG!
SELECT full_name FROM teams WHERE state = 'Texas'  -- RIGHT
```

### ❌ Using == instead of =
```sql
WHERE pts == 20  -- SQL uses single =
WHERE pts = 20   -- RIGHT
```

### ❌ Forgetting to filter by games played
```sql
-- BAD: Includes players with only 1-2 games
SELECT pts FROM player_season_stats WHERE pts >= 20

-- GOOD: Only meaningful playing time
SELECT pts FROM player_season_stats WHERE pts >= 20 AND gp >= 40
```
