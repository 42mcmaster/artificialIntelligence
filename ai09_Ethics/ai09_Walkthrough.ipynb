## Sub-Lesson 09a — AI Ethics & Bias

### Real-World Case: Amazon Hiring AI (2018)

Amazon built an AI system to automate resume screening. The system was trained on 10 years of historical hiring data. In the tech industry during that period, approximately 60% of new hires were men.

**What happened?**
The AI learned patterns from the training data and began penalizing women's resumes. It would downrank applications containing the word "women's" (e.g., "women's chess club") and auto-reject candidates from all-women's colleges.

**Why?**
- The training data reflected the gender imbalance already present in tech hiring
- The model learned: "male candidate" → "likely hire"
- The model learned: "female candidate" → "likely reject"
- The AI was making decisions based on BIAS in the training data, not on actual qualifications

**Lesson:** Biased data creates biased AI. If your training data reflects historical discrimination, your model will learn to discriminate.

### Real-World Case: Healthcare Algorithm (2019)

A hospital system deployed an algorithm to allocate scarce medical resources (hospital beds, specialists, etc.). The algorithm was trained on historical cost data — specifically, how much each patient had spent on healthcare in the past.

**What happened?**
The algorithm systematically gave BLACK PATIENTS lower care recommendations than white patients with identical conditions.

**Why?**
Black Americans historically have less access to healthcare and spend less money on it (due to systemic barriers, not choice). The algorithm learned: "low healthcare cost" → "less medical need." It didn't directly consider race — but it learned a BIASED PROXY VARIABLE that correlated with race.

**Lesson:** Discrimination can happen even when you don't explicitly use protected attributes (race, gender). Using correlated variables can be just as harmful.

### Real-World Case: Facial Recognition Disparities

Facial recognition systems tested on diverse faces show concerning patterns:
- Error rate on darker-skinned women: 10-40%
- Error rate on lighter-skinned men: <1%

**Why?**
Most facial recognition training data comes from Western sources, which skew toward lighter skin tones and European features. When tested on faces not represented in training data, the system fails.

**Real consequence:** In 2020, a Black man was arrested based on a false facial recognition match. The system misidentified him.

**Lesson:** Test your models on DIVERSE data. If your training data is not representative, your model will fail on underrepresented groups.

### Real-World Case: Predictive Policing (Self-Fulfilling Prophecy)

Police departments use AI to predict which neighborhoods will have the most crime, so they can assign officers there.

**The problem:** Training data = past arrest records.
Past arrest records reflect where police have been deployed — not necessarily where actual crime happens.

**What happens:**
1. AI trained on historical arrests
2. AI predicts high crime in over-policed neighborhoods (because arrests are concentrated there)
3. Police deploy MORE officers to those neighborhoods
4. More officers find more arrests
5. Next year's training data shows even MORE arrests in those neighborhoods
6. Cycle repeats

The AI prediction BECOMES SELF-FULFILLING: It's not predicting crime — it's predicting where police will look for crime.

**Lesson:** Understand how predictions affect behavior. AI systems can create feedback loops that reinforce bias.

### How Bias Enters AI Systems

**Bias Source 1: Biased Training Data**
- Historical data reflects past discrimination
- Underrepresented groups have fewer training examples
- Labels themselves might be biased (e.g., "loan approval" reflects historical lending discrimination)

**Bias Source 2: Proxy Variables**
- Using variables correlated with protected attributes
- Example: ZIP code correlates with race, can lead to housing discrimination

**Bias Source 3: Design Choices**
- What features to include? What features to exclude?
- How to measure "success"? Different metrics can hide bias
- These choices reflect the values and blindspots of the development team

**Bias Source 4: Feedback Loops**
- AI predictions change how people behave
- New data reflects both actual patterns AND response to predictions
- Bias gets amplified over time

### Ethical Frameworks for AI

When building AI systems, four ethical principles guide decision-making:

**1. Fairness**
- AI should not discriminate based on protected attributes (race, gender, religion, age, etc.)
- All groups should receive equal treatment
- Different groups should have similar error rates

**2. Transparency**
- Users should understand:
  - What data the AI uses
  - How decisions are made
  - When they're interacting with AI vs. humans
  - The limitations of the system

**3. Accountability**
- Someone must be responsible when AI causes harm
- There should be mechanisms for people to appeal AI decisions
- Organizations should conduct audits and test for bias

**4. Explainability**
- Humans should be able to understand WHY the AI made a specific prediction
- Black box models (like deep neural networks) make this harder
- Sometimes transparency and accuracy trade off against each other

### High-Stakes Domains

Some domains are more critical than others. Errors in high-stakes domains cause real harm.

**HIRING**
- AI decides who gets a job interview
- Impact: Economic livelihood, career, family stability
- Risk: Perpetuating discrimination against women, minorities, older workers

**HEALTHCARE**
- AI diagnoses diseases, recommends treatments
- Impact: Life and death decisions
- Risk: Systematic under-treatment of marginalized groups

**CRIMINAL JUSTICE**
- AI predicts recidivism (likelihood to re-offend)
- AI determines bail amounts, sentence length
- Impact: Freedom, time in prison, family separation
- Risk: Racial bias in arrest records leads to racial bias in predictions

**LENDING**
- AI approves credit, sets interest rates
- Impact: Economic opportunity, wealth building
- Risk: Redlining (loan discrimination by geography/demographics)

### Reflection Question

**Who is responsible when an AI system causes harm?**

Consider: An AI hiring system penalizes women, and a woman is denied a job.

Who is at fault?
- The data scientists who trained the model?
- The engineers who deployed it?
- The hiring managers who used it?
- The company that built it?
- The company that used it?

The honest answer: **Everyone in the chain has some responsibility.**

- Data scientists: Should have tested for bias
- Engineers: Should have included safeguards
- Hiring managers: Should have questioned the AI's recommendations
- Companies: Should have conducted fairness audits

This is why ethics and responsibility matter in AI development.

## Sub-Lesson 09b — Data Privacy Laws

### Overview: Why Privacy Laws Exist

AI systems process MASSIVE amounts of personal data:
- Names, addresses, phone numbers
- Medical histories, diagnoses, prescriptions
- Student grades, test scores, disciplinary records
- Bank accounts, credit card numbers, transaction history
- Browsing history, search history, location data

If this data is stolen, leaked, or misused, it can harm millions of people.

**Privacy laws exist to:**
1. Limit what data organizations can collect
2. Require consent before using data
3. Set security standards to prevent theft
4. Give people rights over their own data
5. Punish organizations that violate privacy

As developers, you MUST understand which laws apply to YOUR data.

### HIPAA: Health Insurance Portability and Accountability Act (1996)

**What it protects:** Protected Health Information (PHI)

**PHI examples:**
- Patient name
- Social Security number
- Date of birth
- Medical record numbers
- Diagnosis codes ("Patient has diabetes")
- Prescription information
- Test results

**Who it applies to:**
- Hospitals and healthcare providers
- Health insurance companies
- Healthcare clearinghouses
- Any organization handling patient health data

**Key requirements:**
- Obtain patient consent BEFORE using health information
- Encrypt health data at rest and in transit
- If breached, notify patients within 60 days
- Implement physical, technical, and administrative safeguards
- Limit data access to employees who need it

**Violations have steep penalties:** $100 - $50,000 per violation, millions in total fines

### HIPAA Violation Example

**Scenario:** A healthcare AI startup trains a model to predict which patients are at high risk for readmission. They download a dataset of 10,000 patient records from a hospital (names, ages, diagnoses, medications, outcomes).

**Problems:**
1. Did they get patient consent? Probably not.
2. Is the data encrypted? Probably not (it's just a CSV file).
3. If the data is leaked, can patients be notified? Hard to do without a name→email mapping.
4. The startup is not a covered entity under HIPAA, but they're handling PHI illegally.

**Consequence:** Fines, lawsuits, shutdown of the project.

### FERPA: Family Educational Rights and Privacy Act (1974)

**What it protects:** Student education records

**Protected information includes:**
- Student name and ID
- Date of birth
- Social Security number
- GPA, test scores, grades
- Course enrollment
- Disciplinary records
- Special education information
- Teacher evaluation notes

**Who it applies to:**
- Public schools (K-12)
- Universities and colleges
- Educational agencies
- Institutions receiving federal funding

**Key requirements:**
- Students (18+) and parents have the right to see their records
- Schools cannot disclose records without consent
- Schools must notify students/parents of FERPA rights
- Exceptions: Directory information (with opt-out), legitimate educational interests

**Violations:** Loss of federal funding, lawsuits, reputational damage

### FERPA Violation Examples

**Example 1:** A teacher posts student test scores on a public bulletin board with names visible.
- Violation: Public disclosure without consent
- Even posting anonymously with student ID numbers can be a violation

**Example 2:** A university builds an AI to predict student dropout risk. They train it on student records (grades, course enrollment, demographics) WITHOUT informed consent.
- Violation: Disclosing educational records without consent
- The students didn't know their data was being used for AI training

**Example 3:** A school district shares a student's special education records with a third-party AI company for "research purposes."
- Violation: Disclosure to unauthorized third party
- Research purposes are not automatic exceptions to FERPA

### GDPR: General Data Protection Regulation (2018, EU)

**What it protects:** Personal data of EU residents (applies WORLDWIDE if they're processing EU data)

**Examples of personal data:**
- Name, email, phone number
- IP address, cookies, browsing history
- Location data
- Biometric data (face, fingerprints)
- Bank account number
- Purchase history
- ANY data that identifies a person

**Who it applies to:**
- Any organization processing data of EU residents
- European companies
- US tech companies (Google, Facebook, Amazon)
- Hospitals, schools, e-commerce sites anywhere in the world (if they have EU users)

**Key principles:**
1. **Lawfulness, fairness, transparency** — Must have legal basis for processing
2. **Purpose limitation** — Data used only for stated purpose
3. **Data minimization** — Collect only what you need
4. **Accuracy** — Keep data correct and up-to-date
5. **Storage limitation** — Don't keep data longer than needed
6. **Integrity and confidentiality** — Secure storage and encryption

### GDPR Rights

Under GDPR, people have explicit rights over their data:

**Right to Access**
- You can ask a company: "What data do you have about me?"
- They must provide it in a readable format (usually within 30 days)

**Right to Rectification**
- You can ask to correct errors
- Example: "Your system says I'm 45, but I'm 35. Fix it."

**Right to Erasure ("Right to be Forgotten")**
- You can ask a company to DELETE your data
- Exception: They can keep it for legal/regulatory reasons
- Example: Delete my profile from your website

**Right to Restrict Processing**
- You can ask a company to stop processing your data (but keep storing it)
- Example: "Stop using my data for ad targeting"

**Right to Data Portability**
- You can ask for your data in a portable format (JSON, CSV)
- Allows you to move to a competitor
- Example: Export all your Facebook data

**Right to Object**
- You can opt out of certain processing
- Example: "Don't use my data for marketing"

**VIOLATIONS:** Fines up to €20 million or 4% of annual revenue (whichever is higher)

### GDPR Violation Example

**Scenario:** A US e-commerce company (Amazon, eBay) collects data from EU customers. Every customer gets added to an advertising database without explicit consent.

**Problems:**
1. No legal basis for processing (should have explicit opt-in consent)
2. Data minimization violated (collecting email, phone, address when not needed for shipping)
3. No privacy policy explaining data use
4. No way to opt out
5. Data shared with third-party ad networks (unauthorized recipients)

**Consequence:** 
- Facebook was fined €5 billion (2019)
- Google was fined €1.5 billion (2022)
- Fines are ongoing for GDPR violations

### PCI DSS: Payment Card Industry Data Security Standard (2004)

**What it protects:** Payment card data (credit cards, debit cards)

**Protected information:**
- Card number (PAN: Primary Account Number)
- Card verification code (CVV/CVC)
- Cardholder name
- Expiration date
- Card holder address

**Who it applies to:**
- Credit card processors
- Merchants/retailers
- Payment gateways (Stripe, PayPal, Square)
- Banks and financial institutions
- ANYONE storing, processing, or transmitting card data

**Key requirements:**
- Encrypt card data at rest and in transit (TLS/SSL)
- Never store full card number + CVV together
- Regular security audits and penetration testing
- Access control (only employees who need it)
- Vulnerability scanning and patching
- Maintain audit logs of all card data access

**VIOLATION:** Fines $5,000 - $100,000 per month, potential liability for fraud losses

### PCI DSS Violation Examples

**Example 1:** A small e-commerce store stores credit card numbers in a plain text file on their server.
- Violation: No encryption
- Consequence: If hacked, customer data is exposed

**Example 2:** A payment processor stores full credit card number + CVV in the same database.
- Violation: Unnecessary sensitive data storage
- If one field is exposed, both are exposed
- Better: Store only last 4 digits and use encrypted tokens

**Example 3:** A company processes millions in card payments but doesn't conduct security audits.
- Violation: No vulnerability scanning
- Consequence: Hackers find and exploit security holes

### Data Privacy Concepts

**Data Minimization**
- Collect only the data you actually need
- More data = larger attack surface
- Example: Need email for password reset? Don't also collect phone number, address, age

**Informed Consent**
- Users must understand what data you collect and how you use it
- Consent must be freely given (not required to access service, unless necessary)
- Privacy policies should be readable, not 50 pages of legal jargon

**Breach Notification**
- If data is stolen or lost, you must notify affected people
- Timeline varies by law (HIPAA: 60 days; GDPR: varies by situation)
- Notification should include: what data was exposed, when, what are the risks

**Right to Deletion**
- GDPR and similar laws give people the right to ask for data deletion
- Must be technically feasible
- Exceptions: Legal hold, regulatory requirement

**Confidentiality, Integrity, Availability (CIA Triad)**
- **Confidentiality:** Keep data secret (encryption, access controls)
- **Integrity:** Ensure data isn't tampered with (checksums, digital signatures)
- **Availability:** Ensure data is accessible when needed (backups, redundancy)

### How Privacy Laws Apply to AI

AI systems collect and process massive amounts of personal data:

**During Training:**
- Need consent to use data for training
- Must comply with data minimization
- Data must be encrypted
- Access must be limited to necessary personnel

**During Deployment:**
- Predictions might reveal sensitive information
- Example: A credit score model that reveals someone's financial situation
- Must give people right to explanation (some jurisdictions)

**Data Retention:**
- How long to keep training data?
- GDPR: Shouldn't keep indefinitely
- Must implement data deletion mechanisms

**Third-party Sharing:**
- If you use cloud services (AWS, Google Cloud), you're sharing data with third parties
- Must have data processing agreements
- Still your responsibility if data is breached

**Regulatory Risk:**
- Building AI without considering privacy = lawsuit, fines, reputation damage
- Privacy should be built in from the start ("privacy by design")
