## Part 2: How It Works (Example Answer)

**Q1. Describe the AI system.**

ChatGPT is a large language model (LLM) made by OpenAI. It's an AI that generates human-like text based on prompts. Users interact with it through a chat interface. It's used for:
- Answering questions
- Writing essays/code
- Customer service
- Content creation
- Education
- Research

The problem it solves: Automating text generation and answering questions at scale. Makes information more accessible.

**Why chosen:** ChatGPT is widely used by students and professionals. Understanding its ethics/limitations is relevant.

**Q2. What data does it collect?**

User inputs:
- Chat messages (questions, prompts)
- User account info (email, account settings)
- Usage patterns (what topics users ask about, how often)
- Metadata (timestamps, IP address, device type)

According to OpenAI's privacy policy, they use this data for:
- Improving the model
- Security and fraud prevention
- Service improvement

Some user data may be retained for content moderation (checking for abuse).

**Q3. What's in the training data?**

ChatGPT was trained on:
- Publicly available text from the internet (websites, books, articles, social media)
- Data cutoff: April 2023 (doesn't know recent events)
- Data from Common Crawl, books, academic papers, Wikipedia, etc.

Note: The exact composition is not fully public, but OpenAI has been transparent that it includes:
- News articles (which have their own biases)
- Social media (biased toward vocal users, often toxic)
- Wikipedia (volunteer-written, may reflect editor biases)
- Academic papers (published authors, skewed toward certain perspectives)

The model was fine-tuned using:
- Human feedback (RLHF — Reinforcement Learning from Human Feedback)
- Annotators (likely mostly from wealthy countries)
- Feedback reflecting their values and biases

## Part 3: Bias Analysis (Example Answer)

**Q4. Could this AI have bias?**

YES, multiple sources of bias:

1. **Training data bias:**
   - Internet text over-represents English speakers, wealthy countries
   - Under-represents voices from Africa, Latin America, Asia, Middle East
   - Social media training data reflects social media biases (toxicity, polarization, sensationalism)
   - Wikipedia reflects volunteer editor biases (men are 90% of editors)

2. **Labeling bias (RLHF):**
   - Human annotators rating "good" responses reflect their cultural values
   - Mostly annotators from wealthy countries
   - OpenAI's values embedded in the fine-tuning process
   - Western/liberal values may be over-represented

3. **Proxy variables:**
   - Model learns subtle patterns about demographics from text
   - Example: "Doctor" appears more often with "he"; "nurse" with "she"
   - Model reproduces these patterns in outputs

4. **Missing perspectives:**
   - Knowledge of non-English cultures/languages is limited
   - Underrepresented perspectives (indigenous, minority, Global South)
   - Model is biased toward dominant narratives

**Q5. Consequences if biased?**

1. **Perpetuating stereotypes:**
   - Student asks: "What do people from [country] believe?"
   - ChatGPT responds with stereotypes from training data
   - Student assumes AI answer is factual
   - Bias → Misinformation → Harm to stereotyped groups

2. **Reinforcing gender/racial biases:**
   - "Programmer" → Gets example of man
   - "Nurse" → Gets example of woman
   - "Scientist" → Over-represents white people
   - Students internalize these biases

3. **Exclusion of non-English speakers:**
   - Model is weaker in non-English languages
   - Non-English speaking students get worse answers
   - Reinforces linguistic colonialism (English dominance)

4. **Policy recommendations:**
   - If used for policy/decision-making, could harm minorities
   - Example: "What's the best immigration policy?" → Biased answer

5. **Erosion of trust:**
   - If marginalized groups realize the model is biased against them, they won't trust it
   - Widens gap in AI access/trust

**Q6. How to test for bias?**

1. **Gender bias testing:**
   - Prompt: "Write a story about a [job]. What gender is the main character?"
   - Test multiple jobs (doctor, nurse, engineer, teacher, etc.)
   - Measure: What percentage are female? Does it match reality?
   - Example: Doctors in real world are ~37% female in US. If ChatGPT says 15%, it's biased.

2. **Racial bias testing:**
   - Prompt: "Describe [country's culture/people]"
   - Does it use stereotypes or nuanced descriptions?
   - Compare to fact-checking sources
   - Measure: Stereotype frequency vs. neutral descriptions

3. **Language bias:**
   - Test same question in English vs. other languages
   - Compare quality of answers
   - Measure accuracy separately by language

4. **Demographic prompt testing:**
   - Ask it to write sentences about people of different demographics
   - Do stereotypes appear?
   - Does model assume race/gender when not mentioned?

5. **Red-teaming:**
   - Try to make the model produce biased output
   - Document where it fails
   - Share results with developers

6. **Fairness metrics:**
   - Measure demographic parity (equal outcomes across groups)
   - Measure equalized odds (equal false positive/negative rates)
   - But note: Different fairness metrics may conflict (no perfect solution)

## Part 4: Privacy Analysis (Example Answer)

**Q7. What personal data does it handle?**

Direct:
- Chat messages (user's questions/prompts — can contain personal information)
- Email address (account creation)
- Payment info (if paying for Plus subscription)
- Browsing data (if using ChatGPT plugins)

Indirect:
- IP address
- Device type/OS
- Timestamps
- API usage logs (if used via API)
- Conversation metadata (length, language, topics)

**Q8. Which privacy laws apply?**

1. **GDPR (EU):** If user is from EU or OpenAI processes EU data
   - Applies to EU user accounts
   - Requires consent, transparency, right to deletion

2. **CCPA (California):** If user is from California
   - Requires disclosure, opt-out of data sales
   - Right to know what data is collected

3. **CFAA (Computer Fraud and Abuse Act, US):** If data is obtained illegally
   - Applies to terms of service violations

4. **State privacy laws:** Various state laws (New York, Colorado, etc.)

5. **NO explicit HIPAA/FERPA requirement** because OpenAI isn't a healthcare provider or school
   - BUT: If users input health/student data, those laws MAY apply
   - Example: Student asks ChatGPT about medical condition → Student's health info is disclosed to OpenAI

**Q9. Privacy risks?**

1. **Accidental data disclosure in responses:**
   - User asks: "Can you summarize my medical record?" (pastes doctor notes)
   - ChatGPT remembers this in context
   - Later, if OpenAI uses chat data for training, medical info could leak
   - Risk: Others might be able to infer sensitive info from model outputs

2. **Data retention:**
   - OpenAI retains chat data for 30 days (default)
   - Longer for paid users
   - Not clear exactly what happens after 30 days
   - Risk: Data could be used for training without clear notice

3. **Data sharing:**
   - OpenAI may use chat data to improve the model
   - Users not always informed
   - Risk: Personal conversations become training data

4. **Third-party access:**
   - OpenAI uses cloud infrastructure (Azure)
   - Data goes through Microsoft servers
   - Risk: More parties have access than user realizes

5. **Lack of transparency:**
   - Privacy policy is unclear about AI training on chat data
   - Opt-out options limited
   - Risk: Users don't know how their data is used

6. **Inference attacks:**
   - Model trained on data could "remember" individuals
   - Researchers can sometimes extract training data from ML models
   - Risk: Even if user data deleted, it could be recovered

7. **School/enterprise use:**
   - Schools use ChatGPT in classrooms
   - Student data sent to OpenAI servers
   - Risk: FERPA violation if student data not properly consented

**Q10. Current safeguards?**

1. **Positive (in place):**
   - HTTPS encryption in transit
   - Default 30-day chat deletion
   - Privacy policy posted
   - Opt-out option for data training (opt for ChatGPT, not for API)
   - Content policy against illegal activity
   - No GDPR opt-out for EU (working on compliance, but not full GDPR implementation)

2. **Gaps (missing):**
   - No easy way to verify data was deleted
   - No control over what data is used for training
   - Limited transparency on data retention after "deletion"
   - No granular consent (can't consent to storage but not training)
   - API users can't opt out of data training
   - No automatic de-identification before training
   - Weak controls on accidental disclosure in responses (model still outputs real names sometimes)

## Part 5: Ethical Considerations (Example Answer)

**Q11. Ethical framework:**

**Fairness:**
- NOT fair to all groups
- Underrepresents non-English perspectives
- Reproduces gender/racial biases
- Less helpful for non-Western users
- Recommendation: Test for demographic fairness; include diverse training data; audit outputs for bias

**Transparency:**
- PARTIALLY transparent
- Positive: OpenAI publishes technical papers, safety reports
- Negative: Training data composition not fully disclosed; model limitations not always clear; fine-tuning process opaque
- Users don't realize model is trained on their data
- Recommendation: Clearer disclaimers; explicit opt-in for data training; publish fairness audits

**Accountability:**
- WEAK accountability
- OpenAI responsible for harms, but:
  - How to sue? Can't sue AI itself
  - OpenAI's terms of service limit liability
  - Users blamed for misuse ("don't trust everything ChatGPT says")
  - No clear appeals process if output is biased
- Recommendation: Liability clarity; appeals process for harmful outputs; transparency reports

**Explainability:**
- LOW explainability
- LLMs are black boxes (we don't know why model generated specific output)
- Can't tell if bias or genuine knowledge
- No feature importance or decision path
- Recommendation: Research interpretability; include confidence scores; flag when making uncertain claims

## Part 6: Proposed Safeguards (Example Answer)

**Q12. Safeguards during development:**

1. **Data collection:**
   - Include more non-English text (proportional to world population)
   - Actively seek out underrepresented perspectives
   - Filter out biased/toxic training data
   - Weight diverse sources equally (don't overweight majority voices)

2. **Fairness testing:**
   - Test for gender/racial/cultural bias before release
   - Measure accuracy separately by language
   - Benchmark against diverse user populations
   - Red-team for bias (try to make model produce biased output)

3. **Diverse team:**
   - Include people from Global South, non-English speakers
   - Include women, minorities in AI development
   - Diverse perspectives catch blind spots
   - Include ethicists, not just engineers

4. **Documentation:**
   - Document training data sources and known limitations
   - Be clear about what the model can/can't do
   - Include uncertainty quantification (model should say "I'm not sure")

**Q13. Safeguards during deployment:**

1. **Transparency:**
   - Clear disclaimer: "This is an AI. It makes mistakes. Don't trust everything it says."
   - Explain how training data could introduce bias
   - Show limitations for specific use cases
   - Tell users their data is used for training (if true)

2. **User control:**
   - Easy opt-out from data training
   - Opt-in (not opt-out) for using chat data to improve model
   - User data deletion that's actually implemented
   - Verification that deletion worked

3. **Monitoring:**
   - Track reports of biased outputs
   - Monitor for harms (discrimination claims, misinformation spread)
   - Regular bias audits (quarterly, not yearly)
   - Public transparency reports on harms discovered

4. **Restrictions on high-stakes use:**
   - Educational use: Warn students not to rely solely on ChatGPT
   - Healthcare: Block medical diagnosis requests (too risky)
   - Legal advice: Warn it's not a lawyer
   - Hiring: Don't use for candidate screening (too biased)

5. **Human-in-the-loop:**
   - Require human review for sensitive topics
   - Don't fully automate decision-making (writing ok, hiring not ok)

**Q14. Organizational/regulatory:**

1. **OpenAI's responsibility:**
   - Publish annual fairness audits
   - Establish clear liability for harms
   - Create appeals process for biased outputs
   - Invest in interpretability research
   - Pay for data (compensate sources)

2. **Regulation:**
   - Require disclosure of training data sources
   - Mandatory bias testing before release
   - GDPR compliance (right to deletion, opt-out)
   - Liability standards (OpenAI can't disclaim all responsibility)
   - Restrictions on high-stakes use cases

3. **Oversight:**
   - External audits (not just OpenAI's internal)
   - Academic researchers get access for bias testing
   - User representatives on AI safety board
   - Whistleblower protections for employees reporting harms

4. **Industry standards:**
   - AI companies share best practices for fairness/privacy
   - Public datasets for bias testing
   - Standardized fairness metrics

## Part 7: Reflection (Example Answer)

**Q15. Overall assessment:**

**Should this AI exist?**
YES. ChatGPT is genuinely useful:
- Makes information accessible
- Assists learning, writing, coding
- Democratizes access to knowledge
- Innovation is valuable

BUT: Not without safeguards. The benefits don't come free.

**Is it currently deployed responsibly?**
PARTIALLY. Good aspects:
- Transparent about limitations (to some degree)
- Not claiming to be perfect
- Attempting to reduce harmful outputs

Problematic aspects:
- Bias issues not fully addressed
- Privacy practices not transparent enough
- Using data for training without full consent
- Being marketed as more capable/reliable than it is
- Users (especially students) don't understand limitations

**Biggest ethical concerns:**
1. **Bias:** Model is trained on English-dominant, Western-biased data. Over-represents some perspectives, under-represents others. Risk of perpetuating stereotypes.

2. **Privacy:** Data used for training without fully informed consent. Users don't realize conversations become training data.

3. **Misuse potential:** Students use it for assignments without understanding it's just predicting text, not understanding. Misinformation gets spread because people trust AI outputs.

4. **Accountability gap:** If someone is harmed by ChatGPT output (follows bad advice, gets discriminated against based on biased output), who's responsible? OpenAI disclaims liability.

**What I'd recommend:**
1. Make bias testing public (show results, not just claims)
2. Real informed consent for data use
3. Restrict use in high-stakes domains (hiring, healthcare, legal)
4. Invest in interpretability (why did it say that?)
5. Clear liability standards
6. Regular external audits

Bottom line: Useful technology, but needs serious guardrails. Don't treat it as truth. Understand its limitations. Hold developers accountable.
