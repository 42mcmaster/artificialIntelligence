## Sub-Lesson 09a — AI Ethics & Bias (Solutions)

### Question 1: How Could Amazon Have Prevented Bias?

**Instructor Answer:**

1. **During Data Collection:**
   - Actively collect diverse historical data
   - Partner with companies that have more balanced hiring
   - Recognize that past data reflects discrimination, not merit
   - Ask: "Does this data represent all qualified applicants?"

2. **During Model Development:**
   - Test the model separately for men and women
   - Measure fairness metrics: Is false rejection rate the same for all groups?
   - Use fairness-aware algorithms that explicitly constrain discrimination
   - Have diverse teams (women, minorities, etc.) audit the model

3. **During Deployment:**
   - Don't use AI alone — have humans review recommendations
   - Require transparency: "Why was this person rejected?"
   - Monitor ongoing decisions for bias drift
   - Give candidates ability to appeal AI decisions

4. **Organizational:**
   - Have clear ethics review process before deploying
   - Test for bias regularly (audits)
   - Be transparent about AI use in hiring

**Key insight:** Preventing bias requires intention. You have to actively look for it.

### Question 2: How to Detect and Fix Healthcare Disparities

**Instructor Answer:**

**Detection:**
1. Disaggregate outcomes by race/ethnicity:
   - Does the model recommend equal care levels for white vs. Black patients with identical conditions?
   - Compare false negative rates (missed diagnoses) by demographic group
   - Compare false positive rates (unnecessary treatments) by demographic group

2. Inspect feature importance:
   - Which variables does the model rely on?
   - Does it use variables correlated with race (like healthcare cost)?

3. Statistical testing:
   - Perform fairness audits (disparate impact analysis)
   - Test if decisions differ significantly across groups

**Fixing it:**
1. **Reframe the problem:** Don't train on healthcare cost. Train on clinical need (symptoms, test results, medical history) only.

2. **Address root causes:** The bias exists because Black Americans face real healthcare barriers (less access, less money, structural racism). Removing race from the model doesn't fix that.

3. **Add fairness constraints:** Train the model with the requirement that error rates must be equal across groups.

4. **Human oversight:** Require doctors to review AI recommendations, especially when they seem unusual.

**Key insight:** Bias often hides in proxy variables. You have to look deeper than just "we didn't use race."**

### Question 3: Predictive Policing — Is It Racist?

**Instructor Answer:**

This is a complex question without a simple answer. Both perspectives are valid:

**Argument that the AI is "racist":**
- It produces decisions that harm Black communities disproportionately
- It perpetuates systemic racism from arrest data
- Result: More police in Black neighborhoods → more arrests → more training data → bias amplifies
- From the perspective of individuals harmed, intent doesn't matter — the outcome is discrimination

**Argument that the problem is systemic:**
- The AI itself isn't inherently racist; arrest data is biased
- The real problem is policing practices and enforcement patterns
- The AI just makes visible an existing bias in the system
- Blaming the AI lets actual human decision-makers off the hook

**Who is responsible?**
- Police who deploy it without understanding limitations
- Data scientists who didn't audit for bias
- Policymakers who allow it without oversight
- Society that accepts biased policing

**Bottom line:** The answer is: both/and. The AI IS producing racist outcomes (harm is real), AND the root problem is systemic racism in policing. You can't fix bias in AI if you don't fix bias in the data and systems it's trained on.

**What should we do?**
- Don't use biased arrest data to train predictive models
- If you must use historical data, acknowledge its bias and test extensively for fairness
- Require human oversight of decisions
- Audit for disparate impact regularly

### Question 4: Privacy-Accuracy Tradeoff

**Instructor Answer:**

**The Problem:**
Adding noise to data for privacy protection (differential privacy) reduces the signal-to-noise ratio, which hurts model accuracy.

**When is accuracy loss acceptable?**

1. **Low-stakes domains:** Recommender systems, ad targeting
   - Impact: User sees different ads/recommendations
   - Acceptable loss: 2-5% accuracy loss for privacy is reasonable
   - Reason: Harm is minimal

2. **Medium-stakes domains:** Student grade prediction, job compatibility
   - Impact: Affects individual opportunities
   - Acceptable loss: 1-2% for privacy
   - Reason: Still make reasonable decisions, and privacy is important

3. **High-stakes domains:** Healthcare diagnosis, criminal justice, lending
   - Impact: Life/death, freedom, economic opportunity
   - Acceptable loss: <1% for privacy
   - Reason: Accuracy is critical. But privacy is still mandatory

**Framework:**
```
Acceptable accuracy loss = (Importance of privacy) / (Criticality of decision)
```

In healthcare: Privacy is important, but a 5% accuracy loss might cost lives. Find solutions that protect both.

**Better approach:** Privacy by design
- Don't collect sensitive data in the first place
- Use federated learning (train models without centralizing data)
- Use secure multi-party computation
- These methods protect privacy without sacrificing much accuracy

### Question 5: What Can You Do?

**Instructor Answer:**

You're an individual developer working at a company. You don't set policy. But you have choices:

**During Data Collection:**
- Ask: "Is this data necessary? Can we collect less?"
- Advocate for diverse data sources
- Flag data quality issues and bias
- Suggest data minimization

**During Model Development:**
- Test for bias and fairness
- Create fairness metrics alongside accuracy metrics
- Include diverse perspectives in code review
- Document assumptions and limitations
- Propose adversarial testing (try to break your own model)

**During Deployment:**
- Suggest A/B testing with humans vs. AI
- Propose logging and monitoring systems
- Build in appeals mechanisms
- Suggest transparency disclosures

**Organizationally:**
- Raise ethical concerns in meetings
- Propose ethics review processes
- Connect with colleagues who care about this
- Document your concerns (email trails matter)

**Worst case:** If the organization won't listen, you have options:
- Whistleblowing (internal and external)
- Leaving the company
- Public statements (if appropriate)
- Supporting advocacy organizations

**Key insight:** You have more power than you think. Companies listen to technical concerns from skilled engineers. Use your voice.

## Sub-Lesson 09b — Data Privacy Laws (Solutions)

### HIPAA Scenario Analysis

**Instructor Note:** These are common scenarios students should recognize as HIPAA violations or compliant.

**Scenario 1: Hospital AI Training**
A hospital wants to train an AI to predict patient readmission risk. They download patient records (names, diagnoses, medications, test results) to a laptop.

**Is this HIPAA compliant?**
- NO. Multiple violations:
  1. Data is unencrypted on a laptop (can be stolen)
  2. Data is on a personal device (access control violation)
  3. Unclear if patients consented to AI use (authorization violation)

**How to fix it:**
- Get patient consent (with clear explanation of AI use)
- Use encrypted, secure servers with access control
- Limit data access to necessary personnel only
- Use anonymized/de-identified data where possible
- Implement audit logs

**Scenario 2: Telehealth App**
A startup builds a symptom checker app. Users enter their medical history and symptoms. The app stores all data in plaintext in a cloud database.

**Is this HIPAA compliant?**
- NO. Violations:
  1. No encryption (confidentiality breach)
  2. No apparent privacy policy or consent
  3. If there's a data breach, no clear breach notification plan
  4. Cloud database — is the provider a Business Associate? (need agreements)

**How to fix it:**
- Encrypt at rest and in transit
- Get explicit informed consent
- Create clear privacy policy
- Use HIPAA-compliant cloud providers
- Implement breach notification plan

**Scenario 3: De-identified Research**
A researcher publishes a study using patient data. They removed names and birthdates. Is this HIPAA compliant?

**Maybe.** HIPAA allows sharing of de-identified data without consent. BUT:
- De-identification is hard (people can be re-identified with other data)
- 18 identifiers must be removed (medical record #, ZIP code, dates, age >89, etc.)
- If any identifier remains, it's still PHI
- Better: Use truly anonymized data or get explicit consent

### FERPA Scenario Analysis

**Scenario 1: Student Grade AI**
A university builds an AI to predict which students are likely to fail and need tutoring. They train it on 10 years of student records (names, grades, course enrollment, major, gender, age).

**Is this FERPA compliant?**
- NO. Violations:
  1. Students probably didn't consent to their data being used for AI training
  2. Data includes potentially sensitive information (gender, age)
  3. Unclear if data is secure or de-identified

**How to fix it:**
- Get student/parent consent BEFORE training
- Use de-identified data (remove names, student IDs)
- Limit data to necessary features (grades, course enrollment, not gender/age)
- Secure storage and encryption
- Transparency: Tell students the AI exists and how it's used

**Scenario 2: Public Bulletin Board**
A school posts honor roll names and GPAs in the hallway.

**Is this FERPA compliant?**
- NO. Disclosure of educational records without consent.
- Even posting first names + grades is a violation
- "Directory information" is an exception, but only if students/parents are notified and given chance to opt out

**How to fix it:**
- Post "honor roll" without GPAs
- Or get explicit family consent
- Or offer opt-out and exclude students who opt out

**Scenario 3: Research at Another University**
A university shares student records with a researcher at another university "for research purposes."

**Is this FERPA compliant?**
- Maybe, but risky. FERPA allows sharing for legitimate educational interest, but:
- Parent/student should be told who has access
- Researcher should have data processing agreement
- Data should be secure
- Better: Get explicit consent

**How to fix it:**
- Get parent/student consent
- Use data processing agreements
- Use de-identified data where possible

### GDPR Scenario Analysis

**Scenario 1: EU User Advertising Database**
A US tech company collects email, name, phone, address, and browsing history from EU users and adds them to an advertising database to target ads.

**Is this GDPR compliant?**
- NO. Multiple violations:
  1. No explicit consent (should be opt-in, not opt-out)
  2. Data minimization: Collecting phone number when not needed
  3. No clear privacy policy explaining ad targeting
  4. No way to opt out
  5. Data shared with ad networks (unauthorized recipients)

**How to fix it:**
- Add clear privacy policy (before collecting data)
- Get explicit opt-in consent
- Collect only email (necessary for targeting)
- Provide right to object/opt-out
- Have data processing agreements with ad networks
- Document consent and retention periods

**Scenario 2: Right to be Forgotten**
An EU user requests: "Delete all my data from your system."
The company says: "We keep data for 7 years for legal reasons, so no."

**Is this acceptable under GDPR?**
- Partially. GDPR allows exceptions for:
  - Legal obligations (tax law, contracts)
  - Legal claims (if there's a lawsuit)
  - But companies must explain why they can't delete
  - Data should be "special restricted purpose" (not used for marketing anymore)

**How to fix it:**
- Explain the legal basis for retention
- Offer alternative: Anonymize instead of delete
- Set retention limits (don't keep forever)
- Have clear retention policies

**Scenario 3: Data Breach Notification**
A company is hacked. Names, emails, and passwords are exposed. They don't tell users until 6 months later.

**Is this GDPR compliant?**
- NO. GDPR requires breach notification "without undue delay" (usually within 72 hours)
- 6 months is way too late
- Users need to know immediately so they can change passwords, monitor for fraud, etc.

**How to fix it:**
- Detect breach quickly
- Notify users within 3 days
- Include: What data was exposed, when, what users should do
- Notify regulators if more than minimal risk

### PCI DSS Scenario Analysis

**Scenario 1: Card Data Storage**
An e-commerce site stores full credit card numbers (1234 5678 9012 3456) in a database.

**Is this PCI DSS compliant?**
- NO. Storing full card numbers is risky and unnecessary
- If the database is breached, all card numbers are exposed
- Modern systems use tokenization: Store only a token, not the card number

**How to fix it:**
- Never store full card numbers
- Use a payment processor (Stripe, PayPal, Square) that handles PCI compliance
- If you must store: Use encryption + tokenization
- Store only last 4 digits (xxxx xxxx xxxx 3456)

**Scenario 2: Unencrypted Payment Transmission**
A website transmits credit card info via HTTP (not HTTPS) to the payment processor.

**Is this PCI DSS compliant?**
- NO. Card data in transit must be encrypted with TLS/SSL
- Unencrypted = hackers can intercept
- All payment pages MUST be HTTPS

**How to fix it:**
- Use HTTPS (TLS 1.2+) for all payment pages
- Use HSTS (HTTP Strict Transport Security) headers
- Disable HTTP entirely
- Regular security audits

**Scenario 3: Employee Access**
A large company stores card data. All 500 employees have access to the database.

**Is this PCI DSS compliant?**
- NO. PCI DSS requires least privilege: Only employees who need access should have it
- 500 employees probably don't all need card data

**How to fix it:**
- Restrict access to payment processing team only
- Use role-based access control
- Log all data access
- Regular audits of who has access
- Multi-factor authentication for high-privilege accounts

## Summary: Laws at a Glance

### When to Use Each Law

| Question | Answer | Law |
|----------|--------|-----|
| Handling patient medical info? | YES → HIPAA | HIPAA |
| Handling student records? | YES → FERPA | FERPA |
| EU residents' personal data? | YES → GDPR | GDPR |
| Handling credit card numbers? | YES → PCI DSS | PCI DSS |
| Tech company, US only? | Probably GDPR (if any EU users) | GDPR |
| Multi-country operation? | All of them (at least) | All |

### Key Differences

**HIPAA vs. FERPA:**
- HIPAA: Health
- FERPA: Education
- Both require consent and have strict penalties

**GDPR vs. US Laws:**
- GDPR: Strongest privacy rights (right to be forgotten, data portability)
- US: Sector-specific (HIPAA for health, FERPA for schools)
- GDPR applies worldwide (if EU residents)

**PCI DSS:**
- Industry standard, not a law (but enforced by card companies)
- Technical focus (encryption, access control)
- High penalties for violations

### Red Flags (Common Violations)

1. **Unencrypted sensitive data** (HIPAA, PCI, GDPR)
2. **No consent/privacy policy** (all laws)
3. **Sharing data without authorization** (all laws)
4. **No breach notification plan** (HIPAA, GDPR, PCI)
5. **No access controls** (all laws)
6. **Keeping data longer than needed** (GDPR, HIPAA)
7. **No data minimization** (GDPR)
8. **Lack of transparency** (GDPR, HIPAA)
