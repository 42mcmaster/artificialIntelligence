## Scenario 1: Hospital AI Training (Solutions)

**Q1a. Which privacy law(s) apply?**

Answer: HIPAA (Health Insurance Portability and Accountability Act)

The hospital is handling Protected Health Information (PHI): names, dates of birth, medical record numbers, diagnosis codes, medications, lab results. HIPAA applies to healthcare providers handling PHI.

**Q1b. Is this scenario COMPLIANT or a VIOLATION?**

Answer: VIOLATION. Multiple problems.

**Q1c. Specific violations:**

1. **Weak encryption/access control:**
   - Password "hospital123" is weak (easy to guess)
   - Shared among 50 people (anyone can access)
   - Excel file stored on a server (not a HIPAA-compliant system)
   - HIPAA requires: Only people who NEED access can have it

2. **No apparent consent documentation:**
   - Unclear if patients consented to their data being used for AI training
   - HIPAA requires authorization before using PHI for research/AI

3. **Data minimization:**
   - Including patient name and DOB unnecessary for AI model
   - Should use de-identified data (remove names, MRNs, DOB)

4. **No audit log:**
   - No record of who accessed the data
   - HIPAA requires logging of all PHI access

**Q1d. How to fix this:**

1. **Get explicit consent:**
   - Have patients sign authorization for use of their data in AI research
   - Explain what the AI will do with their information

2. **De-identify data:**
   - Remove name, date of birth, medical record number
   - Use patient ID numbers instead
   - HIPAA allows sharing of de-identified data without consent

3. **Use secure systems:**
   - Store on HIPAA-compliant servers (AWS HIPAA, Google Cloud HIPAA, on-premise encrypted systems)
   - NOT an Excel file with a weak password
   - Require VPN and multi-factor authentication

4. **Access control:**
   - Only data scientists need access, not 50 people
   - Use role-based access control
   - Log all access attempts

5. **Data retention:**
   - Delete raw data after model is trained
   - Keep only the model (which is de-identified)
   - If keeping raw data, have clear retention policy

## Scenario 2: University Student Records (Solutions)

**Q2a. Which privacy law(s) apply?**

Answer: FERPA (Family Educational Rights and Privacy Act)

The university is handling student education records: grades, GPA, course enrollment, demographic info. FERPA applies to schools and universities.

**Q2b. Does deleting data after 90 days solve the problem?**

Answer: No. Deleting the training data is good (FERPA compliance), but it doesn't address the real violation: Using student data in an AI model WITHOUT their knowledge or consent.

The problem isn't storage — it's disclosure and use without permission.

**Q2c. Is this COMPLIANT or a VIOLATION?**

Answer: VIOLATION of FERPA. Here's why:

1. **No disclosure of AI use:**
   - Students don't know their records are in an AI system
   - FERPA requires transparency: Students have right to know how their data is used

2. **Insufficient consent:**
   - If the university asked, "May we use your records in an AI model to predict dropout risk?" that's consent
   - But if students just received letters saying, "Your records were run through an AI," that's disclosure without prior consent
   - Better: Get explicit opt-in or opt-out before running model

3. **Demographic data use:**
   - Using race/ethnicity and gender in the model could lead to bias
   - Even if not intentional, including these variables is risky and violates principles of privacy protection

**Q2d. How to fix this:**

1. **Get explicit consent:**
   - Send students a message: "We want to use your record to predict dropout risk to help you succeed. Your data will be used only for this purpose and deleted after 90 days. Do you consent?"
   - Allow opt-out
   - Only include students who opt in

2. **De-identify or exclude sensitive data:**
   - Don't include race/ethnicity, gender, or other sensitive attributes
   - Use only: GPA, course enrollment, class year, major

3. **Transparency:**
   - Tell students the model exists
   - Explain how results will be used (advisor outreach, not punishment)
   - Allow access to individual predictions

4. **Data retention:**
   - Delete training data after 90 days (already doing this)
   - Set expiration date for individual predictions (1 year?)
   - Don't keep data longer than needed

5. **Data processor agreement:**
   - If working with external AI company, have a data processing agreement (DPA)
   - Clearly define what they can do with data
   - Require them to delete after specified time

## Scenario 3: EU User Advertising (Solutions)

**Q3a. Which privacy law(s) apply?**

Answer: GDPR (General Data Protection Regulation)

The company collects data from an EU user (Germany). GDPR applies to any organization processing personal data of EU residents, regardless of where the company is located.

**Q3b. Identify all violations:**

1. **Pre-checked consent (not valid):**
   - Pre-checked boxes = user didn't actively agree
   - GDPR requires "freely given" consent (explicit opt-in, not opt-out)
   - Unchecking a box doesn't count as genuine consent

2. **Insufficient transparency:**
   - Privacy policy is 50 pages of legal text
   - Most users won't read it
   - GDPR requires clear, concise, accessible explanations
   - Specific explanation of email marketing needed before collecting data

3. **Excessive data collection:**
   - Collecting browsing history and purchase history
   - For email marketing, need only email address
   - GDPR: Data minimization (collect only what you need)
   - Phone number and address unnecessary for email marketing

4. **Unauthorized data sharing:**
   - Sharing email with ad networks without explicit consent
   - Data processor agreement required with ad networks
   - Users should be told: "We share your data with these partners"

5. **No right to object:**
   - Unclear if user can opt out of marketing
   - GDPR: Right to object to marketing (easy unsubscribe button required)

**Q3c. Is pre-checked consent valid under GDPR?**

Answer: No. GDPR explicitly prohibits pre-checked consent. 

GDPR Article 4 and 7 state:
- Consent must be "freely given"
- User must take an affirmative action (checking a box)
- Pre-checked boxes mean the company is getting consent by default, not free choice
- Silence or inactivity ≠ consent

Fines for this: Up to €20 million or 4% of annual revenue.

**Q3d. How to fix (GDPR compliance):**

1. **Get proper consent:**
   - Remove pre-checked boxes
   - User must ACTIVELY check to opt in
   - Separate consent for email marketing vs. other uses
   - Keep records of consent (when, what they agreed to)

2. **Improve transparency:**
   - Create simple, clear privacy notice
   - Explain in plain language: "We send you emails about products and offers"
   - Explain data sharing: "We share your email with partners to show you ads"
   - Link to full legal policy, but summary should be brief and readable

3. **Data minimization:**
   - For email marketing: Collect ONLY email
   - For targeted ads: Collect email + browsing/purchase history (with consent)
   - Don't collect phone/address unless necessary

4. **Data processor agreements:**
   - Before sharing with ad networks, sign DPA
   - Specify what partners can do with data
   - Ensure partners have appropriate security

5. **Right to object:**
   - Easy unsubscribe link in every email
   - "Manage preferences" page where users control what data is used
   - Honor opt-out requests within 30 days

6. **Data retention:**
   - Don't keep data forever
   - Delete after 2 years if customer inactive
   - Delete immediately if user requests

## Scenario 4: Payment Processing (Solutions)

**Q4a. Which privacy law(s) apply?**

Answer: PCI DSS (Payment Card Industry Data Security Standard)

The store is storing and processing credit card numbers (full card number + CVV). PCI DSS applies to anyone handling card data.

**Q4b. Violations in this scenario:**

1. **Storing full card number + CVV together:**
   - PCI requirement: NEVER store card number + CVV together
   - If database is breached, both values are exposed
   - Standard practice: Tokenize (replace with token), don't store actual card data

2. **Employee direct database access:**
   - Employees shouldn't directly access card data
   - Need system where employees don't see full card numbers
   - Use payment processor API instead (Stripe, PayPal)
   - Process refunds through API, not by manually accessing database

3. **Inadequate access logs:**
   - No mention of audit trail for card data access
   - PCI requires: Every access to card data must be logged
   - Should track: Who, when, what they accessed, what they did

4. **Insufficient security audits:**
   - Only every 2 years is insufficient
   - PCI requires annual penetration testing
   - Large stores: Quarterly (or quarterly on-demand)
   - Small stores (<6M transactions/year): Annual

5. **Storing data unnecessarily:**
   - After processing refund, why keep card data?
   - Should delete after refund processed
   - Or delete after 30 days (industry standard)

**Q4c. The store encrypts card data. Isn't this PCI compliant?**

Answer: No. Encryption is necessary but not sufficient. PCI compliance requires:

1. Encryption of data at rest (✓ they have this)
2. Encryption in transit (✓ HTTPS)
3. Tokenization (storing only tokens, not card data) (✗ missing)
4. Access control (only needed employees) (✗ any employee can access)
5. Audit logging (✗ not mentioned)
6. Regular penetration testing (✗ only every 2 years)
7. Data minimization (✗ keeping unnecessary data)

Just encrypting is like: "I put a lock on the door, so I'm safe." But if 50 people have keys, if the lock is weak, if nobody checks it, if the door is left open — you're still not safe.

**Q4d. How to fix this (PCI compliance):**

1. **Stop storing card data:**
   - Use a payment processor (Stripe, Square, PayPal)
   - They handle PCI compliance
   - Store only Stripe token, not card number
   - When processing refund: Use token to call Stripe API
   - Card data never touches your servers
   - BEST: Use "tokenization" so you never handle raw card data

2. **If you must store card data (avoid this):**
   - Tokenize immediately after authorization
   - Store ONLY token and last 4 digits (xxxx xxxx xxxx 3456)
   - NEVER store CVV
   - Encrypt at rest (AES-256)
   - Encrypt in transit (TLS 1.2+)
   - Restrict database access to 1-2 people (need-to-know)
   - Log all access
   - Delete after 30 days

3. **Access control:**
   - Employees don't access card data directly
   - Use payment API for all card operations
   - Multi-factor authentication for privileged access
   - Separate database account for each person

4. **Security audits:**
   - Annual penetration testing (minimum)
   - Quarterly vulnerability scans
   - Monitor for unauthorized access

5. **Breach notification:**
   - If breached, notify customers, payment processor, card companies
   - Required within specific timeframes
   - Incident response plan

## Scenario 5: Right to be Forgotten (Solutions)

**Q5a. Must the company delete the user's data under GDPR?**

Answer: Partially. The company has LEGAL EXCEPTIONS to the right to be forgotten, but the user's data must still be handled carefully.

**Q5b. Which reasons are valid?**

1. **Tax purposes (VALID):**
   - GDPR allows exceptions for legal obligations
   - Tax law requires keeping financial records for 7 years
   - This is a legitimate legal basis for retention
   - BUT: Data should be restricted to tax purposes only (not used for marketing)

2. **Potential disputes (WEAK):**
   - "Potential" disputes in the future is vague
   - If there's an actual, ongoing dispute, valid
   - But this user has clean history and no current disputes
   - Should have: "User must request in writing; we keep data during dispute"
   - Not: "We'll keep forever in case someday a dispute happens"

3. **Fraud prevention (WEAK):**
   - Fraud prevention is important, but not a blanket exception
   - Should be specific: "Keep data for 2 years for fraud monitoring"
   - Not: "Keep forever in case they commit fraud"
   - Also: If user has clean history, fraud risk is lower

**Q5c. What should the company do?**

Answer: Partial deletion or anonymization:

1. **For tax records (7-year retention):**
   - Keep ONLY: Order date, amount, customer ID
   - DELETE: Name, email, address, payment method
   - Reason: Tax law requires financial records, not full customer profile
   - Outcome: Company can satisfy tax law AND GDPR

2. **For fraud prevention:**
   - Keep: Transaction patterns (anonymized)
   - DELETE: Personal identifying information
   - Offer: Anonymize instead of delete
   - Option: "You have right to deletion. Or we can anonymize (remove name, email, address) — we'd keep only transaction type and amount."

3. **For customer service:**
   - No valid reason to keep personal data
   - If user might open dispute later, that's not a current legal obligation
   - Delete unless there's an active dispute

4. **Best practice:**
   - Send detailed response: "We will delete your personal data (name, email, address, payment method). We will retain only an anonymized transaction ID and amount for tax purposes (required by law for 7 years). We will not use this data for any other purpose."
   - This respects user's right to be forgotten while meeting legal obligations

## Summary Question (Solution)

**Q6. Most common mistake across all scenarios:**

Answer: **Lack of transparency and consent.**

Patterns:

1. **Scenario 1:** No clear consent for AI use of health data
2. **Scenario 2:** Using student data without telling them
3. **Scenario 3:** Pre-checked consent (not real consent)
4. **Scenario 4:** No mention of privacy notice to customers
5. **Scenario 5:** Vague retention policies

Take-home: Organizations often collect data first and ask permission later (or not at all). They assume implied consent. GDPR, HIPAA, FERPA all require:
- Clear notice BEFORE collecting
- Explicit consent (usually opt-in)
- Easy opt-out
- Regular transparency

Second most common: **Data minimization failure.** Collecting more data than needed, storing longer than needed, sharing without purpose.
