## Case 1: Amazon Hiring AI (Solutions)

**Q1a. What was the source of bias in this AI system?**

**Model Answer:**
The training data (2000-2015 hiring decisions) contained a gender imbalance: 60% of hires were men. The AI learned patterns from this biased data and replicated/amplified them. The model couldn't distinguish between "men are actually better at tech" (false) and "Amazon hired more men" (true but biased). Historical discrimination → biased training data → biased model.

**Grading:** Full credit if student mentions:
- Training data was biased/imbalanced
- Model learned from biased historical patterns
- Gender imbalance in hiring created the bias

**Q1b. Who was harmed by this AI system?**

**Model Answer:**
Women applicants to Amazon. They were less likely to be interviewed even if equally qualified. Individual women lost job opportunities. Women collectively were excluded from tech roles. The impact was economic (lost wages) and psychological (discrimination experience). Also: Amazon lost access to female talent.

**Grading:** Full credit if student mentions:
- Women applicants
- Lost job opportunities
- Economic/career harm
- Systemic exclusion

**Q1c. Why didn't Amazon notice the bias earlier?**

**Model Answer:**
- Amazon probably had accuracy metrics (AI screened resumes well), but didn't audit for FAIRNESS
- No one tested for gender bias specifically
- Bias can be hard to notice: The AI wasn't obviously wrong, it just penalized women consistently
- "Accuracy" and "fairness" are different things. High accuracy doesn't mean fair
- Few women in AI/hiring roles to notice the pattern

**Grading:** Full credit if student mentions:
- They focused on accuracy, not fairness
- Lack of explicit bias testing
- Difficulty detecting subtle bias
- Importance of diverse perspectives

**Q1d. If you were building a hiring AI, how would you prevent this bias?**

**Model Answer:**
1. **Data:** Use balanced training data. Partner with companies with diverse hiring. Recognize historical data reflects discrimination.
2. **Testing:** Audit the model separately for men and women. Measure fairness: Same false rejection rate for both groups?
3. **Design:** Have diverse team (women, minorities) review decisions. Use fairness-aware algorithms.
4. **Deployment:** Don't automate fully. Use AI to rank candidates, but have humans make final decisions. Allow appeals.
5. **Monitoring:** Audit ongoing decisions. Track hiring outcomes by gender. Adjust if bias emerges.

**Grading:** Full credit if student mentions:
- Diverse/balanced training data
- Fairness testing (not just accuracy)
- Human oversight
- Diverse team involvement
- Ongoing monitoring

## Case 2: Healthcare Algorithm (Solutions)

**Q2a. Why is using "healthcare cost" as a training variable problematic?**

**Model Answer:**
Healthcare cost is NOT the same as medical need. Black Americans spend less on healthcare due to:
- Lower income/lack of insurance
- Barriers to access (discrimination, geography)
- Systemic healthcare inequality
Using cost → the algorithm inherits this systemic inequality. It trains on a BIASED PROXY: "low cost" becomes "low need," which correlates with race, even though the algorithm never explicitly sees race.

**Grading:** Full credit if student mentions:
- Cost ≠ need
- Systemic healthcare inequality
- Cost is a proxy for race
- How bias enters through "neutral" variables

**Q2b. The algorithm never explicitly used "race" as a variable. So is it racist?**

**Model Answer:**
Yes, in terms of impact. It produced systematically worse outcomes for Black patients, even without explicitly considering race. Intent doesn't matter — harm is real. The mechanism is "disparate impact": a policy that's facially neutral (doesn't mention race) but has a discriminatory effect (worse outcomes for Black people).

However, the root cause isn't the algorithm itself — it's the biased training data reflecting systemic racism in healthcare. You can't build a fair algorithm from unfair data.

**Grading:** Full credit if student mentions:
- Impact matters, not just intent
- Disparate impact even without explicit race
- Root cause is systemic (not algorithm's fault alone)
- Data reflects real-world inequality

**Q2c. How would you detect that this algorithm was producing disparate outcomes?**

**Model Answer:**
1. **Disaggregated analysis:** Compare outcomes (recommended care level) for white vs. Black patients with IDENTICAL medical conditions
2. **Statistical testing:** Is the difference statistically significant? (Not random chance)
3. **Error rates:** Do Black patients have higher false negative rates (missed diagnoses)?
4. **Feature inspection:** Which variables does the model rely on? If using cost, flag it
5. **Fairness metrics:** 
   - Is the false negative rate the same for both groups?
   - Is the error rate the same?
   - Is the recommended treatment level similar?

**Grading:** Full credit if student mentions:
- Comparing outcomes by demographic group
- Controlling for medical condition
- Statistical significance testing
- Error rate analysis
- Feature inspection

**Q2d. What variables SHOULD the algorithm use instead?**

**Model Answer:**
Clinical variables only:
- Symptoms and chief complaint
- Vital signs (BP, heart rate, temperature, oxygen)
- Lab results (blood tests, imaging)
- Diagnosis codes (medical condition)
- Current medications
- Medical history

DO NOT use:
- Cost (biased proxy for healthcare inequality)
- Race (protected attribute + potential for direct discrimination)
- ZIP code (proxy for race)
- Age alone (but relevant medical age-based risks are OK)
- Insurance type (proxy for income/SES)

Use ONLY clinical need, not socioeconomic factors.

**Grading:** Full credit if student mentions:
- Clinical variables (symptoms, vitals, labs)
- Excluding socioeconomic proxies
- Excluding cost/race/geography
- Focusing on medical condition

## Case 3: Facial Recognition (Solutions)

**Q3a. Why is facial recognition accuracy important?**

**Model Answer:**
Facial recognition is used for:
- Criminal investigations (finding suspects)
- Immigration/border control (verifying identity)
- Law enforcement (ID at crime scenes)
- Phone security (unlock with face)

Errors can be catastrophic:
- False positive (wrong person ID'd as criminal) → Innocent person arrested, prosecuted, imprisoned
- False negative (criminal not recognized) → Dangerous person escapes

High stakes = accuracy is CRITICAL. A 5% error rate on facial recognition means thousands of false identifications in a large city.

**Grading:** Full credit if student mentions:
- High-stakes applications (criminal justice)
- Consequences of errors (arrests, prosecution, imprisonment)
- Why accuracy matters

**Q3b. What caused the accuracy disparities in facial recognition?**

**Model Answer:**
Training data bias:
- Most facial recognition datasets come from Western sources (US gov, social media)
- Skewed toward lighter skin tones and European features
- Very few photos of darker-skinned people, especially women
- Model learned patterns on majority population and failed on underrepresented groups
- Underrepresented groups have too few examples for model to learn their features well

Also: Model architecture may have different biases for different features (harder to distinguish among darker-skinned faces if training data had few examples).

**Grading:** Full credit if student mentions:
- Training data imbalance
- Lack of diverse representation
- Model learns majority population better
- Western/light-skinned bias in datasets

**Q3c. If a facial recognition system has 95% overall accuracy, is that good enough for law enforcement?**

**Model Answer:**
No. Here's why:
- 95% accuracy means 5% error rate
- In a city of 1 million, scanning 100,000 faces per day:
  - 5,000 false matches per day
  - 1.8 million false positives per year
- If 10% of people in database are innocent (wrongfully accused of crime), 180,000+ innocent people could be flagged

- Also: 95% average accuracy doesn't mean 95% for all groups. If it's 99% for white men and 80% for Black women (average 95%), that's still problematic

**Better standard:** For criminal justice, need <1% error rate AND error rates equal across demographics.

**Grading:** Full credit if student mentions:
- Mathematical scale (error rate multiplied by population)
- Risk of false positives
- Importance of group-specific accuracy
- Different accuracy for different demographics

**Q3d. How would you test a facial recognition system for bias before deploying it?**

**Model Answer:**
1. **Demographic testing:** Test accuracy separately by:
   - Skin tone (light, medium, dark)
   - Gender (male, female, non-binary)
   - Age (young, middle, old)
   - Combinations (darkest skin + female has highest error rate often)

2. **Fairness metrics:**
   - Is error rate equal across groups?
   - Is false positive rate equal across groups?
   - Is false negative rate equal across groups?
   - What's the worst-case error rate?

3. **Test dataset diversity:**
   - Ensure test data is representative (not just majority)
   - Test on diverse lighting, angles, expressions

4. **Pilot study:**
   - Test on real-world use case before deployment
   - Monitor for errors, especially involving minorities

5. **Documentation:**
   - Be transparent: "System has 15% error rate on dark-skinned women"
   - Include limitations and recommendations

**Grading:** Full credit if student mentions:
- Testing by demographic group
- Measuring fairness across groups
- Worst-case performance
- Real-world testing
- Transparency about limitations

## Case 4: Predictive Policing (Solutions)

**Q4a. Why is arrest data a biased proxy for actual crime?**

**Model Answer:**
Arrest data shows where police LOOKED, not where crime ACTUALLY HAPPENS:
- Police deploy more officers to certain neighborhoods (over-policing)
- More officers → more arrests (even for same crime rates)
- Neighborhoods with bias (over-policed) show higher arrests in data
- But data doesn't show crime in under-policed areas

Example:
- Two neighborhoods, same actual robbery rate
- Neighborhood A (over-policed): 100 robberies, 80 arrests
- Neighborhood B (under-policed): 100 robberies, 20 arrests
- Training data shows: A has more crime (wrong)
- Actual crime rate: Same
- AI learns: Neighborhood A needs more policing

**Grading:** Full credit if student mentions:
- Arrest data ≠ crime data
- Over-policing creates more arrests
- Bias in enforcement shows up in data
- Data doesn't capture under-policed areas

**Q4b. What is a "self-fulfilling prophecy" in the context of predictive policing?**

**Model Answer:**
A self-fulfilling prophecy is when a prediction CAUSES itself to come true:

1. AI predicts: "Neighborhood A has high crime"
2. Police deploy more officers to Neighborhood A
3. More officers find more crime (more arrests, more 911 calls)
4. Next year's data: "Neighborhood A has more crime" (true, but partly because of step 2)
5. AI predicts: "Neighborhood A still has high crime" (even more strongly)
6. Cycle repeats

The prophecy becomes self-fulfilling: The prediction (more police) causes the predicted outcome (more arrests). The original prediction might have been wrong, but police response makes it true.

**Consequence:** Over-policed communities stay over-policed forever, even if actual crime rates don't justify it.

**Grading:** Full credit if student mentions:
- Prediction → action → outcome → reinforces prediction
- Police response causes the predicted outcome
- Cycle amplification
- Perpetuates over-policing

**Q4c. Is the AI system "racist" or is the problem with the data and policing practices?**

**Model Answer:**
Both. It's more accurate to say:
- The AI is NOT inherently racist (it's a tool)
- The AI IS producing racist OUTCOMES (harm is real)
- The ROOT CAUSE is systemic racism in policing (over-policing of Black neighborhoods)
- The AI AMPLIFIES systemic racism by automating and legitimizing biased decisions

Analogy: A biased referee isn't inherently bad, but using a biased referee to make objective decisions makes things worse. The bias was already there; the AI just makes it official.

**Responsibility:** Everyone in the chain:
- Police who deploy predictive policing without understanding bias
- Data scientists who don't audit for disparate impact
- Policymakers who allow it without oversight
- Society that accepts over-policing

You can't fix bias in AI if you don't fix bias in policing.

**Grading:** Full credit if student mentions:
- AI produces racist outcomes
- But root cause is systemic (not just AI)
- AI amplifies/automates existing bias
- Multiple people/institutions responsible

**Q4d. How could you modify a predictive policing system to be more fair?**

**Model Answer:**
1. **Don't use arrest data.** Use:
   - Victim reports (911 calls, crime reports)
   - Actual crime statistics (from crime victims, not arrests)
   - Community surveys
   - Any data source less biased than arrests

2. **Disaggregate analysis:** Test for disparate impact. Does the system recommend more policing in Black neighborhoods even controlling for actual crime?

3. **Community input:** Black communities most affected should have input on the algorithm. Get consent.

4. **Transparency:** Explain which neighborhoods get predicted as high-crime and why.

5. **Regular audits:** Test for bias yearly. If disparate impact emerges, adjust or discontinue.

6. **Human oversight:** Don't automate. Use AI as one input, but require human judgment.

7. **Alternative:** Question whether predictive policing should exist at all. Maybe the issue is policing practices, not the AI.

**Grading:** Full credit if student mentions:
- Use better data (victim reports vs. arrests)
- Test for disparate impact
- Community input and oversight
- Transparency
- Regular audits
- Human oversight or questioning model itself

## Reflection (Solutions)

**Q5. Across all four cases, what is the most common source of bias?**

**Model Answer:**
Training data bias. In ALL four cases:
- Amazon: 60% male in training data
- Healthcare: Cost data reflects systemic inequality
- Facial recognition: Training data skewed toward light skin
- Predictive policing: Arrest data reflects over-policing patterns

The pattern: Biased training data → Biased AI.

Other sources:
- Proxy variables (healthcare cost correlated with race)
- Feedback loops (predictive policing)
- Lack of testing for bias

But the ROOT is: What data you train on determines what patterns the AI learns. Bad data → bad AI.

**Key insight:** Before blaming the AI, audit your data.

**Q6. Who bears responsibility when a biased AI system causes harm?**

**Model Answer:**
Everyone in the chain (though to varying degrees):

1. **Data scientists:** Should test for bias, audit data for historical patterns, refuse to deploy biased models

2. **Engineers:** Should implement fairness monitoring, build in appeals mechanisms, log decisions for audit

3. **Companies/Organizations:** Should have ethics review process, conduct fairness audits, be transparent, limit AI use in high-stakes domains

4. **Managers/Leaders:** Should ask ethical questions, hire diverse teams, fund bias research

5. **Government/Regulators:** Should require bias testing, mandate transparency, punish violations

6. **Society:** Should demand accountability, support victims, advocate for change

**In practice:** Responsibility is SHARED. You can't just blame one person. But as a developer, YOU have some power:
- You can ask hard questions
- You can test for bias
- You can push back on biased designs
- You can document concerns

You're not powerless. Use your voice.
