## Part 3: Build a Prompt from Scratch

You need an AI to help with a real task. Build a complete prompt with all four elements.

```python
# Choose ONE task:
# Option 1: Explain a difficult concept from class (your choice which one)
# Option 2: Generate a lesson plan for something you know about
# Option 3: Create a detailed character for a story
# Option 4: Analyze why a historical event happened

print("PART 3: Build Your Own Prompt")
print()
print("My task: [Describe what you want the AI to do]")
print()
print()

# Now build it with all four elements
context = """
[Your context here: Who is the AI? What's the audience?]
"""
print(f"1. CONTEXT: {context}")
print()

task = """
[Your task here: What exactly should the AI do?]
"""
print(f"2. TASK: {task}")
print()

format_spec = """
[Your format here: How should the output look?]
"""
print(f"3. FORMAT: {format_spec}")
print()

constraints = """
[Your constraints here: Any limits?]
"""
print(f"4. CONSTRAINTS: {constraints}")
print()

print()
print("="*70)
print("COMPLETE PROMPT:")
completePrompt = f"{context.strip()} {task.strip()} {format_spec.strip()} {constraints.strip()}"
print(completePrompt)
```
