## Sub-Lesson 06a — Writing Effective Prompts

### What Makes a Prompt Effective?

A good prompt has four elements:

1. **Context** — Background information the AI needs
2. **Task** — Exactly what you want it to do
3. **Format** — How the response should look
4. **Constraints** — Any limits (length, tone, audience, etc.)

```python
# Let's look at examples of BAD and GOOD prompts

# BAD PROMPT: Too vague, no context, no constraints
badPrompt = "Tell me about machine learning."

print("BAD PROMPT:")
print(f"  {badPrompt}")
print()
print("Problems:")
print("  - No context (Who is this for? What's the background?)")
print("  - No format (Paragraph? Bullet points?)")
print("  - No constraints (How long? How detailed?)")
print("  - Too general (Machine learning is huge!)")
print()
print()

# GOOD PROMPT: Specific context, clear task, format, constraints
goodPrompt = """You are an AI tutor explaining AI to 9th graders. 
Explain machine learning in simple terms using a real-world example.
Use 4-5 sentences. Avoid technical jargon."""

print("GOOD PROMPT:")
print(f"  {goodPrompt}")
print()
print("Why it's better:")
print("  - Context: 'You are an AI tutor explaining to 9th graders'")
print("  - Task: 'Explain machine learning using a real-world example'")
print("  - Format: 'Simple terms, real-world example, avoid jargon'")
print("  - Constraints: '4-5 sentences'")
```

### Prompt Engineering Patterns

Different types of prompts work for different tasks.

```python
# Pattern 1: ZERO-SHOT
# Use when: Quick, simple tasks
# No examples needed

zeroShotPrompt = "What is the capital of France?"

print("ZERO-SHOT PROMPTING")
print("=" * 50)
print(f"Prompt: {zeroShotPrompt}")
print()
print("When to use: Simple questions, quick facts, straightforward tasks")
print()
print()

# Pattern 2: FEW-SHOT
# Use when: Complex or unusual tasks
# Provide examples so the AI understands what you want

fewShotPrompt = """Write a funny product review. Here are examples:

Example 1:
Product: Socks
Review: 'These socks are so soft, I thought I was wearing clouds. 
My feet have filed for independence because they're too happy.'

Example 2:
Product: Water Bottle
Review: 'Holds water. Does it spectacularly? 
Not particularly. But it doesn't leak in my backpack anymore.'

Now write a funny review for: Pencil"""

print("FEW-SHOT PROMPTING")
print("=" * 50)
print(f"Prompt (first 2 lines): {fewShotPrompt[:80]}...")
print()
print("When to use: Complex tasks, unusual outputs, specific styles")
print("Why: Examples show the AI what 'funny' means to you")
print()
print()
```

```python
# Pattern 3: CHAIN-OF-THOUGHT
# Use when: Reasoning, problem-solving, accuracy matters
# Ask the AI to explain step-by-step

chainOfThoughtPrompt = """Solve this problem step-by-step:
'A student buys 3 notebooks at $2.50 each and 4 pens at $0.75 each. 
How much did the student spend total?'

Before giving the final answer, show:
1. What you're calculating
2. The math for each item
3. How you combined them
4. The final answer"""

print("CHAIN-OF-THOUGHT PROMPTING")
print("=" * 50)
print(f"Prompt: {chainOfThoughtPrompt}")
print()
print("When to use: Math problems, logical reasoning, complex analysis")
print("Why: Seeing the steps helps you catch errors and understand reasoning")
print()
print()

# Pattern 4: ROLE-BASED
# Use when: You want a specific perspective, tone, or expertise

roleBasedPrompt = """You are a professional journalist who specializes in technology.
Write a 2-paragraph news article about AI in education.
Use an informative but engaging tone.
Include at least one quote from a fictional expert."""

print("ROLE-BASED PROMPTING")
print("=" * 50)
print(f"Prompt: {roleBasedPrompt}")
print()
print("When to use: Need specific tone, expertise, or perspective")
print("Why: 'You are X' tells the AI how to think and communicate")
```

### Building a Prompt Step-by-Step

Let's build a prompt from scratch using all four elements.

```python
# Let's build a prompt piece by piece

print("BUILDING A PROMPT FROM SCRATCH")
print("=" * 60)
print()

# Start with the task
task = "Create a social media post"
print(f"1. TASK: {task}")
print()

# Add context
context = "You are a marketing manager for a coffee shop called 'Bean There'"
print(f"2. CONTEXT: {context}")
print()

# Specify format
format_req = "The post should be 2-3 sentences, use emojis, and include a call to action"
print(f"3. FORMAT: {format_req}")
print()

# Add constraints
constraints = "Target audience is teenagers and young adults. Tone is casual and fun."
print(f"4. CONSTRAINTS: {constraints}")
print()

# Combine them
completPrompt = f"{context}. {task}. {format_req}. {constraints}"
print("="*60)
print("COMPLETE PROMPT:")
print(completPrompt)
print()
print("This prompt is much better than 'Write a social media post!'")
```

## Sub-Lesson 06b — Evaluating AI Output

### The Five Evaluation Criteria

Whenever you get output from an AI, ask yourself:

```python
# Five ways to evaluate AI output

evaluationCriteria = {
    "Validity": {
        "definition": "Is the output technically correct, logical, and coherent?",
        "example": "If you ask for a math solution, is the math right? Is the explanation clear?",
        "redFlag": "Contradictions, nonsensical statements, broken logic"
    },
    "Relevance": {
        "definition": "Does the output actually answer your question or do what you asked?",
        "example": "If you ask for a poem, did it give you a poem (not an essay)?",
        "redFlag": "Off-topic responses, missing key information you requested"
    },
    "Authenticity": {
        "definition": "Is the output genuinely from the AI or is it copied/plagiarized?",
        "example": "Did the AI generate this, or did it paste from Wikipedia?",
        "redFlag": "Perfect quotes from famous sources, word-for-word matches online"
    },
    "Bias": {
        "definition": "Does the output show unfair stereotypes or discrimination?",
        "example": "If listing jobs, does it only show men as engineers and women as nurses?",
        "redFlag": "Stereotypes, one-sided perspectives, missing diverse viewpoints"
    },
    "Hallucinations": {
        "definition": "Did the AI make up facts that sound real but are actually false?",
        "example": "Fake dates, fake names, fake statistics that sound plausible",
        "redFlag": "Specific 'facts' you can't verify, quotes from people who said them"
    }
}

print("THE FIVE EVALUATION CRITERIA")
print("=" * 70)
for criterion, details in evaluationCriteria.items():
    print(f"\n{criterion.upper()}")
    print(f"  Definition: {details['definition']}")
    print(f"  Example: {details['example']}")
    print(f"  Red Flag: {details['redFlag']}")
```

### Real Examples: Evaluating Output

```python
# EXAMPLE 1: AI Explains Photosynthesis
print("EXAMPLE 1: Evaluating an Explanation")
print("=" * 70)

aiOutput1 = """Photosynthesis is the process where plants turn sunlight into food.
It happens in the chloroplasts of plant cells. The plant takes in carbon dioxide 
from the air and water from the soil. Using sunlight energy, it converts these 
into glucose (sugar) and oxygen. The plant uses the glucose for energy and growth, 
and releases oxygen into the air."""

print("AI Output:")
print(aiOutput1)
print()
print("Evaluation:")
print("  Validity: ✓ Correct and well-explained")
print("  Relevance: ✓ Directly answers 'what is photosynthesis'")
print("  Authenticity: ✓ Paraphrased, sounds natural")
print("  Bias: ✓ No bias detected")
print("  Hallucinations: ✓ All facts are accurate")
print()
print("Overall: GOOD OUTPUT")
print()
print()
```

```python
# EXAMPLE 2: AI Hallucinates Information
print("EXAMPLE 2: Hallucination (Made-Up Facts)")
print("=" * 70)

aiOutput2 = """In 1987, the famous biologist Dr. Margaret Chen discovered that 
plants can hear human voices and respond better to classical music. 
According to her landmark study published in the Journal of Plant Awareness, 
plants grew 40% larger when listening to Mozart. This research changed how 
farmers approach crop cultivation."""

print("AI Output:")
print(aiOutput2)
print()
print("Evaluation:")
print("  Validity: ✗ The facts sound real but... are they?")
print("  Relevance: ~ Talks about photosynthesis indirectly")
print("  Authenticity: ✗ Specific names and dates suggest falseness")
print("  Bias: ✓ No obvious bias")
print("  Hallucinations: ✗ MAJOR PROBLEM!")
print()
print("Red Flags:")
print("  - 'Dr. Margaret Chen' — can you verify this person exists?")
print("  - Specific date (1987) — seems precise, often a hallucination sign")
print("  - 'Journal of Plant Awareness' — does this journal really exist?")
print("  - '40% larger' — exact number with no source")
print()
print("Overall: DO NOT TRUST WITHOUT VERIFICATION")
print()
print()
```

```python
# EXAMPLE 3: AI Shows Bias
print("EXAMPLE 3: Bias in AI Output")
print("=" * 70)

aiOutput3 = """Here are careers in technology:

Engineers: Mostly men, strong in math, work on hardware and software.
UX Designers: Mostly women, good at art, make apps pretty.
Sales: Mostly men, aggressive, close deals.
Administrative: Mostly women, organized, support the team."""

print("AI Output:")
print(aiOutput3)
print()
print("Evaluation:")
print("  Validity: ~ Some accurate info but...")
print("  Relevance: ✓ About tech careers as requested")
print("  Authenticity: ✓ Seems original")
print("  Bias: ✗ MAJOR PROBLEM!")
print("  Hallucinations: ✓ Facts are plausible")
print()
print("Red Flags:")
print("  - Stereotypes about gender and roles")
print("  - 'Mostly men/women' makes assumptions")
print("  - Negative stereotypes ('aggressive' for sales, only 'pretty' for design)")
print("  - Ignores that women can be engineers, men can be designers")
print()
print("Overall: BIASED, SHOULD NOT USE AS-IS")
print("Fix: Rewrite with neutral language and diverse examples")
print()
print()
```

### The Iteration Process

If your first prompt doesn't work, refine it.

```python
# When your output isn't great, iterate:

print("ITERATION PROCESS")
print("=" * 70)
print()

# Step 1: Problem
print("STEP 1: IDENTIFY THE PROBLEM")
firstOutput = "AI is really important in the future."
print(f"  Output: '{firstOutput}'")
print(f"  Problem: Too vague, not specific, not useful\n")

# Step 2: Revise the prompt
print("STEP 2: REVISE THE PROMPT")
originalPrompt = "Tell me about AI in the future."
revisedPrompt = """List 3 specific ways AI will change healthcare by 2030. 
For each, explain: (1) what will change, (2) why it matters, (3) a potential downside. 
Keep each point to 2-3 sentences."""
print(f"  Original: '{originalPrompt}'")
print(f"  Revised: '{revisedPrompt}'\n")

# Step 3: Test and evaluate
print("STEP 3: TEST AND EVALUATE")
print("  Better? Yes! Much more specific.")
print("  Still need improvement? Maybe.")
print()
print("STEP 4: ITERATE AGAIN IF NEEDED")
print("  - Is it too long?")
print("  - Does it have any bias?")
print("  - Are the facts accurate?")
print("  - If 'no' to any, revise again and test")
print()
print("This is NORMAL. Even experts iterate multiple times!")
```
