# ai06a Task: Writing Effective Prompts — SOLUTIONS

**DO NOT DISTRIBUTE TO STUDENTS**

This notebook contains sample solutions for grading.

```python
# SOLUTION: WEAK PROMPT 1
weakPrompt1 = "Tell me about climate change."

print("WEAK PROMPT 1:")
print(f"  {weakPrompt1}")
print()
print("Missing elements: Context (who is the audience?), Format (how long?), Constraints")
print()
print("Sample Improved Version:")
improvedPrompt1 = """You are a science teacher explaining climate change to 9th graders.
Explain the greenhouse effect and one human activity that causes it.
Use 4-5 sentences. Avoid technical jargon. Include one real-world example."""
print(improvedPrompt1)
print()
print("Grading Note:")
print("  Accept any improvement that adds 2+ of the 4 elements.")
print("  Common good versions: Add audience, add length constraint, be more specific")
```

```python
# SOLUTION: WEAK PROMPT 2
weakPrompt2 = "Write a story."

print("WEAK PROMPT 2:")
print(f"  {weakPrompt2}")
print()
print("Missing elements: ALL of them! Context (who?), Task (about what?), Format (length?), Constraints")
print()
print("Sample Improved Version:")
improvedPrompt2 = """You are a creative writing teacher. Write a 3-paragraph adventure story 
for middle school students. The story should: (1) Have a clear beginning, middle, and end, 
(2) Include at least one unexpected plot twist, (3) Use descriptive language.
Keep the reading level appropriate for 12-year-olds."""
print(improvedPrompt2)
print()
print("Grading Note:")
print("  This is the MOST COMMON mistake: 'Write a story' is too vague.")
print("  All 4 elements should be present. If missing 2+, that's a red flag.")
```

```python
# SOLUTION: WEAK PROMPT 3
weakPrompt3 = "List careers in tech."

print("WEAK PROMPT 3:")
print(f"  {weakPrompt3}")
print()
print("Missing elements: Context (for whom?), Format (how many? how detailed?), Constraints (anything specific?)")
print()
print("Sample Improved Version:")
improvedPrompt3 = """List 8-10 different tech careers suitable for high school graduates. 
For each career, include: (1) Job title, (2) Main responsibilities (1 sentence), 
(3) Average salary range. Avoid stereotyping jobs by gender. 
Include diverse roles (not just programming)."""
print(improvedPrompt3)
print()
print("Grading Note:")
print("  Good versions add: Specific number of careers, what details to include,")
print("  and sometimes explicit requirements (no gender stereotypes is BONUS).")
```

## Part 2: Sample Solutions

```python
# SOLUTION: TASK A — World War II Quiz Question
print("TASK A: Generate a World War II quiz question")
print()
print("Pattern choice: Few-shot (because quiz formats benefit from examples)")
print("OR Zero-shot (if student can be clear about what 'quiz question' means)")
print()
quizPrompt = """Write a multiple-choice quiz question about World War II for a 10th grade history class.

Example format:
Question: What year did the United States enter World War II?
A. 1939
B. 1941
C. 1943
D. 1945
Correct answer: B

Now write a new question about: The role of the Holocaust in WWII.
Make it challenging but fair for 10th graders.
Include 4 answer options."""
print(quizPrompt)
print()
print("Grading Notes:")
print("  - Zero-shot is acceptable if student specifies: multiple choice, audience, format")
print("  - Few-shot is excellent because it shows the exact format desired")
print("  - Chain-of-thought would be overkill for this task")
```

```python
# SOLUTION: TASK B — Python Code
print("TASK B: Generate Python code to calculate average")
print()
print("Pattern choice: Chain-of-thought or Zero-shot")
print("OR Few-shot if the student wants to show desired style")
print()
codePrompt = """Write Python code that calculates the average of a list of numbers.
The code should:
1. Ask the user to enter numbers (one at a time, until they type 'done')
2. Store the numbers in a list
3. Calculate and print the average
4. Include comments explaining each section
Keep it simple enough for a beginner programmer."""
print(codePrompt)
print()
print("Alternative (Few-Shot):")
alt = """Show me how to calculate the average of these numbers: [10, 20, 30, 40]
Here's an example of the style I want:

# Set up a list
numbers = [10, 20, 30]
# Calculate the sum
total = sum(numbers)
# Divide by length
average = total / len(numbers)
print(average)

Now write code to ask a user for numbers and calculate their average."""
print(alt)
print()
print("Grading Notes:")
print("  - Accept both zero-shot and few-shot")
print("  - Few-shot shows code style preference (good initiative!)")
print("  - Chain-of-thought is unnecessary for straightforward coding tasks")
```

```python
# SOLUTION: TASK C — Word Problem
print("TASK C: Solve a multi-step word problem")
print()
print("Pattern choice: Chain-of-thought (BEST for reasoning)")
print()
wordProblemPrompt = """Solve this word problem step-by-step:

'Maria is planning a birthday party. She buys 3 boxes of cookies with 24 cookies each. 
She already has 15 cookies at home. She wants to put 6 cookies on each plate for guests. 
How many plates can she fill?'

Before giving your final answer:
1. List what you know (the numbers and what they mean)
2. Show each calculation step
3. Explain your reasoning
4. Give the final answer"""
print(wordProblemPrompt)
print()
print("Grading Notes:")
print("  - Chain-of-thought is STRONGLY preferred for math/logic problems")
print("  - It helps catch errors and verify student understanding")
print("  - Zero-shot would just give an answer without showing work (not good)")
```

## Part 3: Sample Complete Prompt

```python
# SOLUTION: BUILDING A COMPLETE PROMPT
print("SAMPLE SOLUTION — PART 3")
print()
print("Task chosen: Explain the concept of 'friction' in physics")
print()

context = """You are a high school physics teacher explaining a difficult concept.
Your audience is curious 10th graders with basic physics knowledge."""
print(f"1. CONTEXT: {context}")
print()

task = """Explain what friction is and why it matters in everyday life.
Include 3 real-world examples."""
print(f"2. TASK: {task}")
print()

format_spec = """Use clear, simple language with no jargon.
Structure your answer as: 
  - Definition (1-2 sentences)
  - Why it matters (1-2 sentences)
  - Example 1, Example 2, Example 3 (each 1-2 sentences)"""
print(f"3. FORMAT: {format_spec}")
print()

constraints = """Keep total response to about 150 words.
Avoid formulas and math.
Make examples relatable to a 15-year-old's life."""
print(f"4. CONSTRAINTS: {constraints}")
print()

print()
print("="*70)
print("COMPLETE PROMPT:")
completePrompt = f"{context} {task} {format_spec} {constraints}"
print(completePrompt)
print()
print()
print("Grading Rubric for Part 3:")
print("  4 points (Exemplary): All 4 elements present, clear and specific")
print("  3 points (Proficient): 3-4 elements present, mostly clear")
print("  2 points (Developing): 2 elements present or vague wording")
print("  1 point (Beginning): 1 element or very unclear")
```

## Overall Grading Rubric for ai06a_Task

| Component | Developing (1) | Proficient (2) | Advanced (3) | Exemplary (4) |
|-----------|---|---|---|---|
| **Part 1: Identify weak prompts** | Only 1 improved | 2-3 with clear improvements | All 3 improved, good reasoning | All 3 excellent, explains impact |
| **Part 2: Pattern selection** | Vague patterns, poor reasoning | Reasonable patterns, some explanation | Good patterns with clear reasoning | Perfect patterns, explains trade-offs |
| **Part 3: Build from scratch** | 1-2 elements | 3 elements or vague | All 4 elements, mostly clear | All 4 elements, specific & effective |
| **Part 4: Reflection** | Minimal responses | Basic responses | Thoughtful responses | Insightful, shows deep understanding |
| **Overall Quality** | Rushed, many errors | Acceptable, mostly complete | Good, well-organized | Excellent, ready to use |

**Typical Grading:**
- 85-100% = Exemplary/Advanced on most
- 75-84% = Proficient on most
- Below 75% = Significant gaps in understanding

**Teaching Note:** The goal is not perfection—it's understanding that prompts need specificity. Even "Proficient" shows they get the concept.
