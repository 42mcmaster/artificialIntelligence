```python
# SAMPLE SOLUTION: Step 2
print("SAMPLE SOLUTION: STEP 2")
print("="*70)
print()
print("First Prompt:")
print("-" * 70)
firstPrompt = """You are a professional writing assistant. Help me write a polite email 
to my teacher. I need to ask for a 3-day extension on my history project because 
I got sick and missed class. The email should: (1) Apologize for the late request, 
(2) Explain the situation briefly, (3) Ask for the extension, (4) Promise to turn 
it in by the new date. Keep it 3-4 paragraphs. Use a respectful tone."""
print(firstPrompt)
print("-" * 70)
print()
print("My reasoning:")
print("  Context: 'professional writing assistant' sets the right tone")
print("  Task: Clear what I want (email to teacher, asking for extension)")
print("  Format: '3-4 paragraphs, specific order of content'")
print("  Constraints: 'respectful tone, brief explanation'")
```

```python
# SAMPLE SOLUTION: Step 3
print("SAMPLE SOLUTION: STEP 3")
print("="*70)
print()
print("AI Response (simulated):")
print("-" * 70)
aiOutput = """Subject: Request for Extension on History Project

Dear [Teacher Name],

I am writing to request a three-day extension on the history project due [original date]. 
Unfortunately, I was absent from class earlier this week due to illness and fell behind 
on my work.

I understand that this is a short notice and appreciate your consideration. I am fully 
committed to this assignment and believe that the extra time will allow me to produce 
quality work that reflects my understanding of the material.

I will have the completed project submitted by [new date]. Thank you for your time and 
understanding.

Sincerely,
[Your Name]"""
print(aiOutput)
print("-" * 70)
print()
print("Evaluation:")
print()
print("Validity: ✓ Correct")
print("  The email is grammatically correct and logically sound.")
print()
print("Relevance: ✓ Good")
print("  Directly addresses my request. Includes all points I asked for.")
print()
print("Authenticity: ✓ Good")
print("  Paraphrased, sounds like original writing, not copied.")
print()
print("Bias: ✓ Good")
print("  No unfair language or stereotypes.")
print()
print("Hallucinations: ✓ Good")
print("  No false claims. Generic content (no made-up facts).")
print()
print("Main strengths:")
print("  - Professional tone")
print("  - Concise (not too long)")
print("  - Covers all requested points")
print("  - Has email structure (subject, greeting, body, closing)")
print()
print("Main weaknesses:")
print("  - Feels a bit generic/templated")
print("  - Doesn't show much personality")
print("  - Could be slightly more specific about the situation")
print()
print("Overall: GOOD (Would use with minor tweaks)")
```

## Grading Rubric for ai06_DIYTask

| Step | Developing (1) | Proficient (2) | Advanced (3) | Exemplary (4) |
|------|---|---|---|---|
| **Step 1: Challenge Selection** | Unclear or inappropriate | Clear, appropriate | Clear, specific, real | Thoughtful choice with good reasoning |
| **Step 2: First Prompt** | Vague, missing elements | 3 of 4 elements | All 4 elements present | All 4 strong, shows thinking |
| **Step 3: Evaluation** | Only 1-2 criteria | 3-4 criteria, basic | All 5 criteria, fair job | All 5 criteria, thorough |
| **Step 4: Refinement** | No real change | Some improvements | Clear improvements | Strategic changes addressing issues |
| **Step 5: Comparison** | Minimal comparison | Identifies differences | Good before/after analysis | Excellent analysis of impact |
| **Step 6: Reflection** | Minimal answers | Basic answers | Good insight | Deep, thoughtful answers |

**Scoring Guide:**
- 85-100%: 3-4 on most rubrics (Advanced/Exemplary)
- 75-84%: Solid 2s with some 3s (Proficient/Advanced)
- Below 75%: Mostly 1s-2s or incomplete work

**Teaching Notes:**
- Grade holistically—this task shows process, not just product
- A student with poor first output but excellent reflection deserves credit
- If refinement shows NO improvement, that's a flag (they didn't understand iteration)
- Exceptional: Student who tries 3+ iterations on their own
- Look for evidence of actual thinking, not generic answers

**Common Issues:**
- Weak refinement: "I made it longer" (not strategic)
- Vague challenges: "Do something fun" (needs to be specific)
- Skipped evaluation: Just pasted output without analysis
- Generic reflection: Copy-paste answers that don't match their work

**Best Submissions Show:**
1. Real problem that student actually needs to solve
2. Thoughtful first attempt with all 4 elements
3. Honest evaluation that identifies specific problems
4. Strategic revisions addressing actual weaknesses
5. Clear improvement and learning documented
6. Reflection that connects to future use
