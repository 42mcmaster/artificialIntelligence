## Sub-Lesson 06a — Writing Effective Prompts

```python
# SOLUTION: Write Your Own Prompt
# Task: Write a prompt that asks an AI to explain photosynthesis to 7th graders

myPrompt = """You are an AI science tutor explaining difficult concepts to 7th graders.
Explain photosynthesis using simple, everyday language.
Break it down into 3 steps: (1) What goes in, (2) What happens, (3) What comes out.
Avoid scientific jargon. Use 5-6 sentences. Include one analogy (compare it to something familiar)."""

print("SAMPLE SOLUTION: Photosynthesis Prompt")
print("=" * 70)
print(myPrompt)
print()
print("Why this is good:")
print("  Context: 'You are an AI tutor explaining to 7th graders'")
print("  Task: 'Explain photosynthesis'")
print("  Format: 'Simple language, 3 steps, 5-6 sentences, include analogy'")
print("  Constraints: 'Avoid jargon, 7th-grade level'")
print()
print("Teaching Note: Any prompt with these four elements is acceptable.")
print("Students may organize differently, use different analogies.")
print("Grade based on completeness of the four elements, not exact wording.")
```

### Instructor Note: Sub-Lesson 06a Guidance

**Teaching Tips:**
- Emphasize that there's no single "perfect" prompt—context matters
- Have students test prompts in pairs: writer + evaluator
- Show how small changes (adding "in bullet points" or "for a 5th grader") dramatically improve results
- Encourage iteration—celebrate the second/third attempt being better
- Common student mistake: Being too polite ("Could you please") instead of direct

**Common Student Prompts:**
- Too vague: "Write a story" → Better: "Write a 2-paragraph story about a robot learning to feel emotions. Include dialogue. Use simple language."
- Missing audience: "Explain evolution" → Better: "Explain evolution to someone who has never heard of Darwin. Use 4 examples from animals we see today."
- No format: "Tell me about COVID" → Better: "Create a FAQ (Q&A format) about COVID-19 transmission. 5 questions, 1-2 sentence answers each."

## Sub-Lesson 06b — Evaluating AI Output

```python
# SOLUTION: Evaluate This Output
# Identifying all problems in the sample output

sampleOutput = """The best jobs for teenagers are retail and fast food because 
that's what most teens do. Boys are better at working with computers, 
while girls are better at babysitting and tutoring. In 2019, the Bureau 
of Labor Statistics (BLS) reported that 92% of teenage boys prefer tech jobs. 
This is just how people are naturally."""

print("SAMPLE OUTPUT EVALUATION — SOLUTION")
print("=" * 70)
print()

print("VALIDITY:")
print("  Rating: POOR (partially)")
print("  Issues: The statement 'retail and fast food are THE BEST' is opinion, not fact.")
print("          Not all teens do these jobs. Logic is flawed.")
print()

print("RELEVANCE:")
print("  Rating: FAIR")
print("  Comment: Answers the general question about teen jobs, but too narrow.")
print()

print("AUTHENTICITY:")
print("  Rating: QUESTIONABLE")
print("  Issues: The specific '92% of teenage boys prefer tech' sounds made up.")
print("          This level of specific stat should be verified.")
print()

print("BIAS:")
print("  Rating: MAJOR PROBLEMS (strongly biased)")
print("  Issues:")
print("    - Gender stereotypes: 'Boys are better at computers'")
print("    - Gender stereotypes: 'Girls better at babysitting/tutoring'")
print("    - Assumes gender = preference (not true!)")
print("    - Ignores LGBTQ+ identities")
print("    - Implies this is 'natural' (it's not—it's socially constructed)")
print()

print("HALLUCINATIONS:")
print("  Rating: LIKELY PRESENT")
print("  Suspicions:")
print("    - '92% of teenage boys prefer tech jobs' — too specific, unverified")
print("    - Where did '92%' come from? Can't find this stat")
print("    - Sounds like AI invented this statistic")
print()

print("OVERALL ASSESSMENT:")
print("  DO NOT USE THIS OUTPUT.")
print()
print("  Problems:")
print("    1. Contains harmful gender stereotypes")
print("    2. Makes up statistics (hallucination)")
print("    3. Too narrow (only retail/fast food)")
print("    4. Presents opinion as fact")
print()
print("  How to fix:")
print("    - Revise prompt: 'List 10 different jobs teenagers can do. For each,")
print("      explain the benefits. Include jobs across different fields.")
print("    - Avoid gendered language")
print("    - Request fact-based info only")
print("    - Ask AI to cite sources for statistics")
```

### Instructor Note: Sub-Lesson 06b Guidance

**Teaching Tips:**
- Make bias evaluation FUN: Have students find bias in pairs as a game
- Teach students that hallucinations are common—this is NOT the student's fault
- Emphasize: "Always verify factual claims. Never trust AI on dates, statistics, or names without checking."
- Show that hallucinations often have these markers: Specific numbers, proper names, specific dates
- Real-world connection: "Misinformation spreads because people don't evaluate AI output critically"

**Critical Bias Teachable Moment:**
- AI reflects the biases in its training data (mostly Western, mostly English, mostly male-authored text)
- Students should expect to find bias; their job is to catch it and fix it
- This is not about "AI is bad"—it's about "You must be critical consumers"

**Hallucination Reality Check:**
- AI is very confident when lying (doesn't say "I don't know")
- Facts, dates, statistics are particularly prone to hallucination
- One wrong detail doesn't invalidate the whole response—but it needs verification
- If the cost of error is high (medical advice, legal advice), be extra skeptical

## Lesson Wrap-Up Notes

**Key Messages to Reinforce:**

**Real-World Connections:**

- Marketers use AI for copywriting (but have to fact-check claims)

**Common Misconceptions to Correct:**
- "AI always knows the right answer" → NO: AI can hallucinate confidently
- "If an AI says it, it must be true" → NO: Verify important facts
- "Bias in AI is a technical problem" → It's a DATA problem; bias is in training data
- "Good prompts work the same for every AI" → NO: Different models have different strengths
- "Prompt engineering is just asking nicely" → NO: It's strategic, specific communication
