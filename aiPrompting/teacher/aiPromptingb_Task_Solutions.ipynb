## Part 2: Problem Outputs with Solutions

```python
# SOLUTION: PROBLEM OUTPUT 1 (Cats)
problemOutput1 = """The average age of a cat is 15-20 years. Most cats are lazy and just sleep 
all day. Cats are not smart like dogs are. According to Dr. Helen Smith at MIT, 
recent research shows that 95% of cat owners regret getting a cat. This is why 
more people have dogs as pets."""

print("PROBLEM OUTPUT 1 ANALYSIS: About Cats")
print("="*70)
print(problemOutput1)
print()
print()
print("Problems found:")
print()
print("1. BIAS — Unfair comparisons")
print("   'Cats are not smart like dogs are' is a stereotype.")
print("   Cats and dogs have different types of intelligence.")
print()
print("2. HALLUCINATION — Fake source and statistic")
print("   'Dr. Helen Smith at MIT... 95% of cat owners regret getting a cat'")
print("   This is extremely suspicious:")
print("     - The person may not exist")
print("     - MIT doesn't have this research (verifiable)")
print("     - '95%' is absurdly high and unverified")
print("     - Sounds made up to support an opinion")
print()
print("3. FAULTY LOGIC — Unsupported conclusion")
print("   'This is why more people have dogs as pets'")
print("   Even if some people regret cats, this doesn't prove people prefer dogs.")
print()
print("How to fix:")
print("  - Remove the fake study")
print("  - Use neutral comparisons of cat vs dog intelligence")
print("  - Base conclusions on verified statistics")
print("  - Acknowledge that both cats and dogs make good pets")
```

```python
# SOLUTION: PROBLEM OUTPUT 2 (Sandwich)
problemOutput2 = """To make a peanut butter sandwich: First, grab two slices of bread 
and put them on the counter. Next, open the peanut butter jar and spread it on one slice. 
Then, do the same with jelly on the other slice. Push the two slices together. 
Cut diagonally for best results. Serve with a glass of milk and exactly 47 grapes."""

print("PROBLEM OUTPUT 2 ANALYSIS: Recipe Instructions")
print("="*70)
print(problemOutput2)
print()
print()
print("Problems found:")
print()
print("1. HALLUCINATION / ABSURD DETAIL")
print("   'Serve with exactly 47 grapes'")
print("   This is randomly specific and nonsensical.")
print("   No reason for '47' specifically.")
print("   The AI made something up that sounds plausible but is clearly wrong.")
print()
print("2. VALIDITY — Mostly fine, except the grape issue")
print("   The sandwich instructions are logical and accurate.")
print("   The grape detail breaks that validity.")
print()
print("3. IRRELEVANCE — Missing instruction")
print("   Forgot to mention: Should you add jelly to the SANDWICH,")
print("   or is it served on the side?")
print("   (Ambiguous phrasing)")
print()
print("How to fix:")
print("  - REMOVE the '47 grapes' line entirely")
print("  - Clarify jelly application (optional, on sandwich, etc.)")
print("  - Result: Normal, usable recipe")
print()
print("Teaching Note:")
print("  This is a good example of AI sounding confident while being absurd.")
print("  '47 grapes' isn't obviously wrong to everyone—it requires critical thinking.")
```
