- GPA and academic standing

## Scenario 4: Payment Processing

**Context:**
A small online store processes payments using a custom payment form. When customers check out, they enter:
- Full credit card number (16 digits)
- CVV (3-digit security code)
- Card expiration date
- Cardholder name
- Billing address

The data is transmitted over HTTPS (encrypted in transit) to the store's server.

On the server, the full credit card data (number + CVV) is stored in a MySQL database with AES encryption.

When the store processes a refund, an employee logs into the database directly, finds the customer's card number and CVV, and manually processes the refund through the payment processor.

The store conducts security audits every 2 years.
