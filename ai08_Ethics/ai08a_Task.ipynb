## Case 1: Amazon Hiring AI (2018)

**Background:**
Amazon built an automated resume screening system to handle the large volume of job applications. The system was trained on hiring data from 2000-2015, when Amazon was growing rapidly in tech roles. During this period, approximately 60% of new hires in tech were men.

**What Happened:**
In 2015, HR noticed the AI was systematically penalizing female applicants. Examples:
- The AI downranked resumes containing the word "women's" (e.g., "women's chess club")
- Graduates from all-women's colleges received lower scores
- Female candidates with identical qualifications to male candidates scored lower

**Discovery:**
Amazon discovered this when auditing the system. They tried to remove the gender bias, but the model had learned subtle patterns that correlated with gender (e.g., certain word choices, school choices, work history gaps). The model was biased in ways they couldn't easily fix without retraining on different data.

**Outcome:**
Amazon discontinued the AI hiring system (2018).

## Case 2: Healthcare Algorithm Disparities (2019)

**Background:**
A major US hospital system deployed an algorithm to allocate scarce healthcare resources (ICU beds, specialist consultations, expensive treatments). The algorithm was designed to identify which patients needed the most intensive care.

The algorithm was trained on historical cost data — specifically, how much money each patient had spent on healthcare in the past. The logic: "If you've spent a lot, you have serious health needs. If you've spent a little, you're probably healthy."

**What Happened:**
Researchers analyzed the algorithm's decisions and found that BLACK PATIENTS received systematically LOWER care recommendations than white patients with identical medical conditions.

**Why?**
Black Americans historically spend less money on healthcare (due to income inequality, lack of insurance, barriers to care). The algorithm learned: "low healthcare cost" → "low medical need." It didn't directly consider race, but it learned a BIASED PROXY VARIABLE that correlated with race.

**Example:**
Two patients, same heart disease diagnosis:
- White patient: Spent $10,000 on healthcare → Recommended for intensive cardiology care
- Black patient: Spent $3,000 on healthcare → Recommended for basic management

The difference wasn't medical condition (same), it was historical spending.

**Outcome:**
The algorithm was modified, and hospitals now audit algorithms for disparate impact.

## Case 3: Facial Recognition Accuracy (2020s)

**Background:**
Facial recognition systems have become ubiquitous: law enforcement, airport security, phone unlock, social media tagging. Most systems are trained on large publicly available datasets (mostly from Western sources).

**The Problem:**
Research showed significant accuracy disparities:
- Error rate on lighter-skinned men: <1%
- Error rate on darker-skinned women: 10-40%
- Systems were least accurate on darker-skinned females and most accurate on lighter-skinned males

**Why?**
Training data skew: Most facial recognition datasets come from:
- US government databases (NIST, DMV) — majority white
- Social media platforms (Facebook, Instagram) — skewed demographics
- News/media photos — historical media bias
- Very few photos of darker-skinned people, especially women

The model learned patterns on a skewed distribution and failed on underrepresented groups.

**Real Consequences:**
- In 2020, a Black man was arrested in Detroit based on a false facial recognition match (90% error rate for his face)
- UK police facial recognition arrested innocent people
- Studies show it's particularly bad for darker-skinned women

**Outcome:**
Increasing regulation (bans in some cities), calls for better training data, lawsuits against police departments.

## Case 4: Predictive Policing (2010s-2020s)

**Background:**
Police departments across the US have deployed AI systems to predict where crime will occur and help allocate patrol resources.

**How It Works:**
1. Train AI on historical crime data (arrests, 911 calls)
2. AI identifies "hotspots" — neighborhoods predicted to have high crime
3. Police deploy more officers to those neighborhoods
4. More officers find more crime (more arrests)
5. Next year's training data shows more crime in those neighborhoods
6. AI predicts high crime there again
7. Cycle repeats

**The Problem:**
Historical crime data is NOT the same as actual crime. It's a record of where police looked. Over-policed neighborhoods (often Black neighborhoods) have more arrests recorded, even if actual crime rates aren't higher.

The AI learns from biased data and produces a self-fulfilling prophecy:
- Prediction: "High crime in neighborhood A" (actually: high policing in neighborhood A)
- Police deploy more officers
- More arrests (because more police looking)
- Next year's data confirms prediction
- Bias gets worse

**Real Consequences:**
- Over-policed communities become MORE policed
- Young Black men arrested at higher rates (even controlling for actual crime)
- Systemic racism in policing gets automated and amplified

**Outcome:**
Increasing criticism; some cities have limited or ended predictive policing. Academic papers show it amplifies discrimination.
