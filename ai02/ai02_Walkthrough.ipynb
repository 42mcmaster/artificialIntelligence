## Part 1: Decision Tree Concepts (02a)

### What Is a Decision Tree?

A **decision tree** is a flowchart that asks a series of yes/no questions to reach a final answer.

Think of it like a game of "20 Questions":
- Start with a question
- Answer yes or no
- Ask another question based on your answer
- Keep going until you reach the final answer

### Real-World Examples

**Example 1: Jacket Advisor**
- Question 1: Is it cold outside (below 50°F)?
  - YES → Question 2: Very cold (below 30°F)?
    - YES → Wear heavy winter coat
    - NO → Wear light jacket
  - NO → Question 2: Very warm (above 75°F)?
    - YES → Wear shorts
    - NO → Wear regular clothes

**Example 2: Movie Rating Filter**
- Question 1: Do you like action movies?
  - YES → Question 2: Prefer recent movies?
    - YES → Show: *Top Gun Maverick*
    - NO → Show: *Die Hard*
  - NO → Question 2: Do you like comedy?
    - YES → Show: *Knives Out*
    - NO → Show: *Inception*

---

### Anatomy of a Decision Tree

Every decision tree has the same basic parts:

```
             Is it cold? ← ROOT (starting point)
             /        \
           YES        NO ← BRANCHES (paths)
           |          |
        Cold?      Warm? ← NODES (decision points)
        /  \       /  \
      YES  NO    YES  NO ← BRANCHES
      |    |     |    |
     🧥   🧥    👕   👔 ← LEAVES (final answers)
```

| Part | Definition | In Our Example |
|------|-----------|----------------|
| **Root** | The first question you ask | "Is it cold?" |
| **Node** | A decision point (any yes/no question) | "Is it cold?", "Cold?", "Warm?" |
| **Branch** | A path labeled yes/no from one node to another | Each arrow from a decision |
| **Leaf** | The final answer (no more questions) | The clothing recommendations |

### Tracing Through a Tree

To **trace** a tree means to follow the branches based on specific yes/no answers.

**Example: Restaurant Recommender**

```
           Like spicy food?
           /              \
         YES              NO
         |                |
      Budget?          Budget?
      /   \            /   \
    <$15  >$15       <$15  >$15
    |     |          |     |
   Thai  Indian    Italian French
```

**Trace 1: "Yes to spicy, Budget > $15"**
- Start at root: "Like spicy food?" → YES
- Follow left branch to: "Budget?" → >$15
- Follow right branch to leaf: **Indian Palace**

**Trace 2: "No to spicy, Budget < $15"**
- Start at root: "Like spicy food?" → NO
- Follow right branch to: "Budget?" → <$15
- Follow left branch to leaf: **Italian Kitchen**

### When Are Decision Trees Useful?

#### Advantages (Why Use Them?) ✓
- Easy to understand and explain to others
- Works well for categorizing things into groups
- Can be drawn by hand or coded in Python
- Natural way to handle multiple yes/no decisions
- Good for classification problems ("is this A or B?")

#### Limitations (When NOT to Use) ✗
- Only works for yes/no questions (binary decisions)
- Can get very complicated with many branches
- Not good for predicting numbers (e.g., "predict temperature")
- Requires clear categories (doesn't handle "maybe" or "sometimes")
- With huge datasets, random forests are better

**Examples:**
- ✓ Good: "Recommend a movie based on user preferences"
- ✓ Good: "Classify an email as spam or not spam"
- ✗ Bad: "Predict exact stock price next week"
- ✗ Bad: "Estimate how much someone will spend on groceries"

## Part 2: Coding Decision Trees (02b)

Now that you understand tree **concepts**, let's **code** them in Python!

A decision tree in Python is just **nested if/elif/else statements**.

---

### Simple Decision Tree: Jacket Advisor

Let's code the jacket advisor tree from earlier.

````python
# ==========================================
# JACKET ADVISOR: Simple Decision Tree
# ==========================================

# Get temperature input from user
tempInput = input("Enter temperature in Fahrenheit: ")

# Convert string to integer (so we can compare numbers)
temperature = int(tempInput)

# First decision: Is it cold (below 50°F)?
if temperature < 50:
    # YES → Temperature is cold, ask follow-up question
    if temperature < 30:
        # Very cold (below 30°F)
        print("Wear a heavy winter coat")
    else:
        # Just cold (30-50°F)
        print("Wear a light jacket")
else:
    # NO → Temperature is warm or moderate, ask follow-up question
    if temperature > 75:
        # Very warm (above 75°F)
        print("Wear shorts and a t-shirt")
    else:
        # Moderate (50-75°F)
        print("Wear regular clothes")
````

**How it works:**
1. Get temperature from user (as text)
2. Convert to integer so we can compare numbers
3. First `if`: Is it < 50? (cold or not?)
4. Inside that, another `if`: Is it < 30? (very cold?)
5. Each path leads to a different recommendation (a "leaf" in our tree)

### Input Validation: Handling User Mistakes

Users often make mistakes when typing. We need to **validate** their input.

**Common issues:**
- User types "YES" but we check for "yes" (case sensitivity)
- User types "45a" instead of "45" (not a valid number)
- User types extra spaces (" yes " instead of "yes")

````python
# ==========================================
# INPUT VALIDATION EXAMPLE 1: Text Input
# ==========================================

# Problem: User might type "YES" or "yes" or "Yes"
# Solution: Use .lower() to convert to lowercase first

likeActionMovies = input("Do you like action movies? (yes/no): ").lower()
# .lower() converts "YES" → "yes", "Yes" → "yes", etc.

# Now we only check for lowercase "yes"
if likeActionMovies == "yes":
    print("Watch: Top Gun Maverick or Die Hard!")
elif likeActionMovies == "no":
    print("Watch: Knives Out or Inception!")
else:
    # User didn't type yes or no
    print("Please type 'yes' or 'no'")
````

````python
# ==========================================
# INPUT VALIDATION EXAMPLE 2: Number Input
# ==========================================

# Problem: User might type "45a" or "abc" instead of a number
# Solution: Use .isdigit() to check if it's all digits

ageString = input("Enter your age: ")

# Check if the input contains only digits (0-9)
if ageString.isdigit():
    # It's a valid number, convert to integer
    age = int(ageString)
    
    # Now make decisions based on age
    if age < 13:
        print("G and PG movies only")
    elif age < 17:
        print("PG-13 movies allowed")
    else:
        print("R-rated movies allowed")
else:
    # Not a valid number
    print("Error: Please enter a valid number")
````

### Complex Conditions: AND and OR

Sometimes you need to check **multiple conditions** at once.

- **AND**: Both conditions must be true
- **OR**: At least one condition must be true

````python
# ==========================================
# USING AND: Both conditions must be true
# ==========================================

# Example: Decide if we should go to the beach
# Need: Nice weather AND weekend day

temperature = 75  # Degrees Fahrenheit
dayOfWeek = "Saturday"  # What day is it?

# AND: Both must be true
if temperature > 70 and dayOfWeek in ["Saturday", "Sunday"]:
    print("Perfect beach day!")
else:
    print("Not a good time for the beach")
````

````python
# ==========================================
# USING OR: At least one condition is true
# ==========================================

# Example: Cancel school if cold OR snowing
# Don't need both, just one is enough

temperature = 20  # Very cold
isSnowing = False  # Not snowing

# OR: Only one needs to be true
if temperature < 32 or isSnowing:
    print("School is canceled!")
else:
    print("School is open")
````

### More Complex Example: Study Plan Advisor

Let's code a decision tree that recommends a study plan based on hours studied and classes attended.

````python
# ==========================================
# STUDY PLAN ADVISOR: Complex Conditions
# ==========================================

# Get user inputs
studyHoursStr = input("Hours studying per week? ")
classesAttendedStr = input("Classes attended this month? ")

# Validate inputs (check if they're valid numbers)
if studyHoursStr.isdigit() and classesAttendedStr.isdigit():
    # Convert to integers
    studyHours = int(studyHoursStr)
    classesAttended = int(classesAttendedStr)
    
    # Decision tree: Check both study and attendance
    if studyHours >= 10 and classesAttended >= 8:
        # Both good
        print("You're doing great! Keep it up!")
    elif studyHours >= 10 or classesAttended >= 8:
        # One is good, one needs work
        print("Good effort! Try to improve the other one.")
    else:
        # Both need improvement
        print("Increase both study time and class attendance.")
else:
    # Invalid input
    print("Error: Please enter valid numbers")
````
