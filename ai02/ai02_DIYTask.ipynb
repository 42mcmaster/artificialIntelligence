# ai02 DIY Task — Build Your Own Decision Tree Program

## Assignment Overview

This is your **independent project**. You will design and code a **complete decision tree program** on a topic of your choice.

### Requirements

Your program must:
1. **Ask 3+ questions** (to create 8+ possible outcomes)
2. **Use nested if/elif/else** properly
3. **Validate all inputs** (using `.lower()`, `.isdigit()`, or checking valid options)
4. **Handle errors gracefully** (clear error messages)
5. **Have comments on every line** explaining the logic
6. **Use camelCase** for all variable names
7. **Be tested thoroughly** (at least 8 test cases)
8. **Include a write-up** explaining your design
