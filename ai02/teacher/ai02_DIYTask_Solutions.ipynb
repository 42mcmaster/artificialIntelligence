## Sample Solution: Career Path Advisor

This is an example of a complete DIY project to show students what "excellent" looks like.

````python3
# ==========================================
# DIY DECISION TREE: Career Path Advisor
# ==========================================

# Welcome message
print("Welcome to the Career Path Advisor!")
print("Answer a few questions to get your personalized recommendation.\n")

# Question 1: Work environment preference
workEnvironment = input("Do you prefer working with people or with technology? (people/technology): ").lower()

# Validate question 1
if workEnvironment not in ["people", "technology"]:
    print("Error: Please type 'people' or 'technology'")
else:
    # Question 2: Income importance
    incomeImportant = input("Is high income important to you? (yes/no): ").lower()
    
    # Validate question 2
    if incomeImportant not in ["yes", "no"]:
        print("Error: Please type 'yes' or 'no'")
    else:
        # Question 3: Creativity vs. Structure
        creativity = input("Do you prefer creative/flexible work or structured/routine? (creative/structured): ").lower()
        
        # Validate question 3
        if creativity not in ["creative", "structured"]:
            print("Error: Please type 'creative' or 'structured'")
        else:
            # Decision tree with 8 outcomes
            if workEnvironment == "people":
                # Works with people
                if incomeImportant == "yes":
                    # People + High Income
                    if creativity == "creative":
                        # People + Income + Creative
                        print("\nRecommended Career: Marketing Executive")
                        print("Why: High income, work with clients/teams, creative campaigns")
                    else:
                        # People + Income + Structured
                        print("\nRecommended Career: Healthcare Manager")
                        print("Why: High income, manage medical staff, structured procedures")
                else:
                    # People + Not High Income
                    if creativity == "creative":
                        # People + Not Income + Creative
                        print("\nRecommended Career: Art Teacher")
                        print("Why: Work with students, creative expression, moderate income")
                    else:
                        # People + Not Income + Structured
                        print("\nRecommended Career: Counselor or Social Worker")
                        print("Why: Help people, follow established procedures, meaningful work")
            else:
                # Works with technology
                if incomeImportant == "yes":
                    # Technology + High Income
                    if creativity == "creative":
                        # Tech + Income + Creative
                        print("\nRecommended Career: Game Developer")
                        print("Why: Creative coding, high income potential, cutting-edge tech")
                    else:
                        # Tech + Income + Structured
                        print("\nRecommended Career: Software Engineer at Large Company")
                        print("Why: High income, structured processes, technology focus")
                else:
                    # Technology + Not High Income
                    if creativity == "creative":
                        # Tech + Not Income + Creative
                        print("\nRecommended Career: Web Designer")
                        print("Why: Create beautiful interfaces, moderate income, creative tech work")
                    else:
                        # Tech + Not Income + Structured
                        print("\nRecommended Career: IT Support Specialist")
                        print("Why: Help with tech issues, structured troubleshooting, stable work")
````

### What Makes This Solution Excellent

✓ **3 questions asked** in logical order
✓ **8 different outcomes** (2^3 = 8 paths)
✓ **Proper nesting**: Each question asked before follow-up
✓ **Input validation**: Checks all inputs, clear error messages
✓ **Comments**: Every section explained
✓ **camelCase variables**: `workEnvironment`, `incomeImportant`, `creativity`
✓ **Rich output**: Shows career AND reasoning
✓ **Testable**: Easy to verify all 8 paths work
