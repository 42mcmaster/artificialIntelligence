## Task 1: Restaurant Recommender (Full Solution)

```python3
# ==========================================
# Task 1: Restaurant Recommender (Solution)
# ==========================================

# Ask first question and convert to lowercase
spicyChoice = input("Do you like spicy food? (yes/no): ").lower()

# Validate first input
if spicyChoice not in ["yes", "no"]:
    print("Error: Please type 'yes' or 'no'")
else:
    # Ask second question and convert to lowercase
    budgetChoice = input("Is your budget less than $15? (yes/no): ").lower()
    
    # Validate second input
    if budgetChoice not in ["yes", "no"]:
        print("Error: Please type 'yes' or 'no'")
    else:
        # Decision tree: nested if statements
        if spicyChoice == "yes":
            # User likes spicy, check budget
            if budgetChoice == "yes":
                # Spicy AND budget-friendly
                print("Recommended: Thai Bistro")
            else:
                # Spicy AND willing to spend more
                print("Recommended: Indian Palace")
        else:
            # User doesn't like spicy, check budget
            if budgetChoice == "yes":
                # Not spicy AND budget-friendly
                print("Recommended: Italian Kitchen")
            else:
                # Not spicy AND willing to spend more
                print("Recommended: French Bistro")
```

### Instructor Note
Key teaching points:
- Nested if: Budget question is only asked after spicy answer
- Input validation: Check before using input
- `.lower()`: Handles "YES", "Yes", "yes" uniformly
- Four distinct paths lead to four different recommendations
- Error handling for invalid input (first check)

## Task 2: Custom Decision Tree (Sample Solutions)

### Option A: Movie Recommender

```python3
# ==========================================
# Custom Tree Option A: Movie Recommender
# ==========================================

# Question 1: Genre preference
likeAction = input("Do you like action movies? (yes/no): ").lower()

# Validate first question
if likeAction not in ["yes", "no"]:
    print("Error: Please type 'yes' or 'no'")
else:
    # Question 2: Mood
    likeComedy = input("Do you want something funny or serious? (funny/serious): ").lower()
    
    # Validate second question
    if likeComedy not in ["funny", "serious"]:
        print("Error: Please type 'funny' or 'serious'")
    else:
        # Decision tree
        if likeAction == "yes":
            # User likes action
            if likeComedy == "funny":
                # Action AND funny
                print("Recommended: Mission Impossible (fun action)")
            else:
                # Action AND serious
                print("Recommended: The Dark Knight (serious action)")
        else:
            # User doesn't like action
            if likeComedy == "funny":
                # Drama AND funny
                print("Recommended: The Grand Budapest Hotel (quirky humor)")
            else:
                # Drama AND serious
                print("Recommended: Inception (thoughtful sci-fi)")
```

### Option B: Workout Recommender

```python3
# ==========================================
# Custom Tree Option B: Workout Recommender
# ==========================================

# Question 1: Time available
timeStr = input("Do you have at least 30 minutes? (yes/no): ").lower()

# Validate time question
if timeStr not in ["yes", "no"]:
    print("Error: Please type 'yes' or 'no'")
else:
    # Question 2: Experience level
    experienceStr = input("Are you experienced with exercise? (yes/no): ").lower()
    
    # Validate experience question
    if experienceStr not in ["yes", "no"]:
        print("Error: Please type 'yes' or 'no'")
    else:
        # Decision tree
        if timeStr == "yes":
            # 30+ minutes available
            if experienceStr == "yes":
                # Experienced AND lots of time
                print("Recommended: Full gym workout (45 min total body)")
            else:
                # Beginner AND lots of time
                print("Recommended: Easy jog + stretching (30 min)")
        else:
            # Less than 30 minutes
            if experienceStr == "yes":
                # Experienced BUT short on time
                print("Recommended: Quick strength training (15-20 min)")
            else:
                # Beginner AND short on time
                print("Recommended: Quick walk or light cardio (15 min)")
```

### Option C: What to Eat Recommender

```python3
# ==========================================
# Custom Tree Option C: What to Eat
# ==========================================

# Question 1: Budget
budgetStr = input("Is your budget under $10? (yes/no): ").lower()

# Validate budget question
if budgetStr not in ["yes", "no"]:
    print("Error: Please type 'yes' or 'no'")
else:
    # Question 2: Time
    timeStr = input("Do you have at least 15 minutes? (yes/no): ").lower()
    
    # Validate time question
    if timeStr not in ["yes", "no"]:
        print("Error: Please type 'yes' or 'no'")
    else:
        # Decision tree
        if budgetStr == "yes":
            # Limited budget
            if timeStr == "yes":
                # Budget AND time to cook
                print("Eat: Homemade pasta with tomato sauce")
            else:
                # Budget AND quick
                print("Eat: Instant ramen or cereal")
        else:
            # More budget
            if timeStr == "yes":
                # Money AND time
                print("Eat: Order from a nice restaurant")
            else:
                # Money BUT no time
                print("Eat: Fast food (burger, sandwich)")
```

## Common Student Mistakes & How to Address

### Mistake 1: No Input Validation
**Problem:** Code assumes user always types "yes" or "no"
**Fix:** Always use `.lower()` and check if input is in valid list

### Mistake 2: Wrong Indentation
**Problem:** Nested if statements at wrong level
**Fix:** Review indentation rules; use IDE that shows indent guides

### Mistake 3: Using `=` Instead of `==`
**Problem:** `if choice = "yes":` assigns instead of compares
**Fix:** Use `==` for comparison, `=` only for assignment

### Mistake 4: Duplicate Code
**Problem:** Same validation logic repeated many times
**Fix:** Consider using a loop to ask until valid input, or functions (advanced)

### Mistake 5: Unclear Variable Names
**Problem:** Variables like `x`, `y`, or `a`, `b`
**Fix:** Use descriptive camelCase: `userChoice`, `budgetAmount`, `timeAvailable`
