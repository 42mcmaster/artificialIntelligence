## Scenario 1: Video Game AI Behavior

### Task 1a: Draw the Tree (Sample Answer)

```
          Is player within 10m?
          /                   \
        YES                   NO
        |                     |
    Health > 50%?         Ammo in area?
    /         \           /         \
  YES        NO        YES         NO
  |          |         |           |
 Chase     Flee     Search      Patrol
```

### Task 1b: Trace These Paths (Answer Key)

**Scenario A:** Player is 5 meters away, NPC has 60% health
- Path: Player nearby? **YES** | Health > 50%? **YES**
- Outcome: **Chase the player** ✓

**Scenario B:** Player is 15 meters away, there is ammo nearby
- Path: Player nearby? **NO** | Ammo available? **YES**
- Outcome: **Search for ammo** ✓

**Scenario C:** Player is 8 meters away, NPC has 30% health
- Path: Player nearby? **YES** | Health > 50%? **NO**
- Outcome: **Run away** ✓

### Task 1c: Identify Tree Parts (Answer Key)

- **Root**: "Is player within 10m?" ✓
- **Total Nodes**: 3 nodes (root + "Health > 50%?" + "Ammo in area?") ✓
- **Total Leaves**: 4 outcomes (Chase, Flee, Search, Patrol) ✓
- **Longest Path**: 2 questions (always pick one from first decision, then one follow-up) ✓

### Instructor Note
This is a well-balanced binary tree. Students should recognize that:
- Every path requires exactly 2 decisions before reaching a leaf
- All 4 paths are equally viable
- This design makes game AI responsive but not overly complex

## Scenario 3: Email Spam Filter

### Task 3a: Brainstorm Questions (Sample Answers)

Good questions might include:
1. **"Does it contain suspicious keywords (free money, click here, verify)?"** ✓
2. **"Is it from a known contact or trusted organization?"** ✓
3. **"Does it contain suspicious links or attachments?"** ✓
4. **"Is the sender's email address properly formatted?"** ✓
5. **"Are there spelling/grammar errors?"** ✓

### Instructor Note
Accept any reasonable yes/no questions about email characteristics. Less useful questions:
- "Is it in English?" (too broad)
- "Is it long?" (not relevant to spam)
- "Does it have capital letters?" (not a strong indicator)

Encourage students to think about real spam signals.

### Task 3b: Design Your Tree (Sample Answer)

```
              From trusted contact?
              /                   \
            YES                   NO
            |                     |
     Contains suspicious        Contains suspicious
     keywords/links?            keywords/links?
     /           \               /           \
   YES          NO            YES            NO
   |            |             |              |
SUSPICIOUS  LEGITIMATE    DEFINITELY     SUSPICIOUS
                           SPAM
```

**Logic:**
- Trusted contact + safe = **Legitimate** (deliver to inbox)
- Trusted contact + suspicious = **Suspicious** (move to spam, but notify user)
- Unknown sender + suspicious = **Definitely Spam** (block)
- Unknown sender + safe = **Suspicious** (user hasn't seen before, so be cautious)

### Task 3c: Test Your Tree (Sample Answers)

**Email 1:** "Congratulations! You won $1 million! Click here to claim."
- Questions asked: **From trusted contact? NO** → **Contains suspicious keywords? YES** ("won", "click here")
- Classification: **_X_ Definitely Spam** ✓

**Email 2:** "Project meeting rescheduled to 2pm Thursday. - Manager Sarah"
- Questions asked: **From trusted contact? YES** (manager at work) → **Contains suspicious keywords? NO**
- Classification: **_X_ Legitimate** ✓

**Email 3:** "Hi, click this link to verify your password urgently"
- Questions asked: **From trusted contact? NO** (phishing attempt) → **Contains suspicious keywords? YES** ("verify password", "urgently", "click link")
- Classification: **_X_ Definitely Spam** ✓

### Instructor Note
Good student answers should show clear logic. Some students might classify Email 3 differently depending on whether they recognize it as phishing. That's fine—the important thing is that they follow their tree's logic consistently.

## Reflection Questions (Sample Answers)

### 1. Why use decision trees in Scenario 1?

**Sample answer:**
Decision trees are good for video game AI because:
- NPCs need to make quick decisions (if/else is fast)
- Behavior is based on simple yes/no factors (player distance, health %)
- Different outcomes are clearly defined (Chase, Flee, etc.)
- Easy to adjust/test (change thresholds like "10m" or "50%")

When they'd NOT be good:
- If we needed to predict continuous values ("NPC should move 45.3 pixels") → use neural networks instead
- If behavior needs "fuzzy" logic ("somewhat aggressive") → need probability/fuzzy logic

### 2. Advantages of the college major tree?

**Sample answer:**
- **Personalized**: The tree asks about YOUR interests, not generic
- **Conversational**: Easier than reading a wall of text with all majors
- **Guided**: Narrows down options progressively instead of overwhelming
- **Easy to understand**: Anyone can follow the logic
- **Repeatable**: If answers change, you can trace a new path

### 3. Decisions that are HARD to represent as trees?

**Sample answers:**
- "What job should I take?" - Involves many factors (salary, location, happiness, growth) that don't fit cleanly into yes/no
- "Will I like this movie?" - Too many subtle factors, not binary decisions
- "What's the best investment?" - Requires numerical prediction, probability
- "How happy are you?" - Continuous spectrum, not discrete categories

**Key point**: Trees work best when decisions are **binary** (yes/no), **clear**, and **independent**.
