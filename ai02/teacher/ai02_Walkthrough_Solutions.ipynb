## Part 1: Decision Tree Concepts (02a)

### Label the Parts (Answer Key)

Tree to label:
```
           Do you like pizza?
           /              \
         YES              NO
         |                |
      Cheese?         Vegetables?
      /   \            /     \
    YES   NO         YES     NO
    |     |          |       |
  Pepperoni Veggie Salad   Soup
```

**Answers:**
1. **Root**: "Do you like pizza?" (the first/top question)
2. **Number of nodes**: 3 total ("Do you like pizza?", "Cheese?", "Vegetables?")
3. **Number of branches**: 6 total (each yes/no path from each node)
4. **Leaves (final answers)**: Pepperoni, Veggie, Salad, Soup

### Instructor Note
Students sometimes confuse "branches" with "leaf nodes." Clarify that branches are the **paths**, while leaves are the **final answers**. A tree with 3 nodes typically has 6 branches (each node has a yes and no path).

### Try This: Test the Jacket Advisor (Answer Key)

Run the jacket advisor code with these inputs:

1. Temperature = 25°F → Output: **"Wear a heavy winter coat"** ✓
   - 25 < 50 (yes, cold) → 25 < 30 (yes, very cold) → heavy coat

2. Temperature = 45°F → Output: **"Wear a light jacket"** ✓

3. Temperature = 60°F → Output: **"Wear regular clothes"** ✓
   - 60 < 50 (no, not cold) → 60 > 75 (no, not very warm) → regular clothes

4. Temperature = 80°F → Output: **"Wear shorts and a t-shirt"** ✓
   - 80 < 50 (no, not cold) → 80 > 75 (yes, very warm) → shorts and tee

### Instructor Note
Use these tests to verify students understand nested if logic. Ask them to trace their own temperature values to check understanding.

### Try This: Add Validation (Answer Key)

Jacket advisor with input validation:

````python3
# ==========================================
# JACKET ADVISOR WITH VALIDATION
# ==========================================

# Get temperature input from user
tempInput = input("Enter temperature in Fahrenheit: ")

# Validate input: Check if it's a valid number
if tempInput.isdigit():
    # Convert string to integer
    temperature = int(tempInput)
    
    # First decision: Is it cold (below 50°F)?
    if temperature < 50:
        # YES → Temperature is cold, ask follow-up question
        if temperature < 30:
            # Very cold (below 30°F)
            print("Wear a heavy winter coat")
        else:
            # Just cold (30-50°F)
            print("Wear a light jacket")
    else:
        # NO → Temperature is warm or moderate, ask follow-up question
        if temperature > 75:
            # Very warm (above 75°F)
            print("Wear shorts and a t-shirt")
        else:
            # Moderate (50-75°F)
            print("Wear regular clothes")
else:
    # Invalid input (not a number)
    print("Error: Please enter a valid number")
````

### Instructor Note
Key changes:
1. Wrap entire logic in `if tempInput.isdigit()`
2. Add `else` clause for invalid input
3. Indentation is critical—everything after `if` is indented one level

Common student mistakes:
- Forgetting `.isdigit()` check
- Wrong indentation
- Using `int()` before validation (will crash if input is "abc")

### Try This: Restaurant Recommender (Answer Key)

Complete solution with validation and nested conditions:

````python3
# ==========================================
# RESTAURANT RECOMMENDER: Complete Solution
# ==========================================

# Question 1: Do you like spicy food?
spicyChoice = input("Do you like spicy food? (yes/no): ").lower()

# Validate: Only accept "yes" or "no"
if spicyChoice not in ["yes", "no"]:
    print("Error: Please type 'yes' or 'no'")
else:
    # Question 2: Is your budget less than $15?
    budgetChoice = input("Is your budget less than $15? (yes/no): ").lower()
    
    # Validate second input
    if budgetChoice not in ["yes", "no"]:
        print("Error: Please type 'yes' or 'no'")
    else:
        # Decision tree: Spicy? Budget?
        if spicyChoice == "yes":
            # Likes spicy food
            if budgetChoice == "yes":
                # Spicy + Budget
                print("Recommended: Thai Bistro")
            else:
                # Spicy + Expensive
                print("Recommended: Indian Palace")
        else:
            # Does not like spicy
            if budgetChoice == "yes":
                # Not spicy + Budget
                print("Recommended: Italian Kitchen")
            else:
                # Not spicy + Expensive
                print("Recommended: French Bistro")
````

### Instructor Note
Key teaching points:
1. **Nested if**: The budget question is asked inside the spicy answer
2. **Validation with `not in`**: Check if input is valid before proceeding
3. **Four outcomes**: Spicy+Budget, Spicy+Expensive, NotSpicy+Budget, NotSpicy+Expensive
4. **Indentation levels**: Shows structure clearly

Alternative approach (using `and`):
```python
if spicyChoice == "yes" and budgetChoice == "yes":
    print("Thai Bistro")
elif spicyChoice == "yes" and budgetChoice == "no":
    print("Indian Palace")
# ... etc
```
Both approaches work; choose based on student preference.

## Teaching Tips

### Common Errors & Fixes

| Error | Why | Fix |
|-------|-----|-----|
| `if temperature = 50:` | Single `=` assigns, not compares | Use `==` for comparison |
| String comparison issues | "YES" ≠ "yes" (case matters) | Use `.lower()` first |
| IndentationError | Python requires consistent indentation | Use 4 spaces per level |
| `int(input())` with non-number | Crashes if user enters "abc" | Check `.isdigit()` first |
| Logic error in branches | Wrong condition chosen | Trace manually with examples |

### Debugging Strategy

When a student's code doesn't work:
1. **Trace manually**: Follow their code with an example input
2. **Check indentation**: Use Python's error messages to find misaligned blocks
3. **Add print statements**: Insert `print()` to see which branch was taken
4. **Validate step by step**: Check each input and condition separately

### Assessment Ideas

- Have students draw a tree diagram for their code
- Ask them to trace 3-4 different paths through their tree
- Have them explain why `.lower()` or `.isdigit()` is needed
- Challenge: Add a third question to expand the tree
