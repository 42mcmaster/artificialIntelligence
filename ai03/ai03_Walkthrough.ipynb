## Step 4: Data Cleaning

Real-world data is messy! We need to clean it before training a model.

**Cleaning steps**:
1. **Select relevant columns**: Choose only the features we need for prediction
2. **Handle missing values**: Remove rows with NaN (missing data)
3. **Encode categorical data**: Convert text to numbers (one-hot encoding)

### Why do we do this?
- Machine learning models need numeric data
- Models fail with missing values
- Irrelevant columns add noise and slow down training

````python
# STEP 4A: Select only the columns we need
# We'll use: pclass, survived, sex, age, sibsp, parch, fare, embarked
# We drop: name, ticket, cabin, boat, body, home.dest (not useful for prediction)

columnsToKeep = ['pclass', 'survived', 'sex', 'age', 'sibsp', 'parch', 'fare', 'embarked']
titanicData = titanicData[columnsToKeep]  # Keep only these columns

print("✓ Selected relevant columns")
print(f"Dataset shape after column selection: {titanicData.shape}")
print(f"Columns: {list(titanicData.columns)}")
````

### Why we dropped certain columns:

- **name**: Individual names are unique, not predictive
- **ticket**: Ticket numbers are identifiers, not features
- **cabin**: Almost all missing values (263 out of 1309)
- **boat**: Only has values for survivors (would be cheating!)
- **body**: Only has values for non-survivors (would be cheating!)
- **home.dest**: Too many missing values, not essential

We keep the others because they directly relate to survival!

````python
# STEP 4B: Check for missing values
print("\nMissing values BEFORE cleaning:")
print(titanicData.isnull().sum())  # Count missing values in each column
print(f"\nTotal missing values: {titanicData.isnull().sum().sum()}")
````

### Instructor Note

Notice that `age` and `embarked` have missing values:
- age: Missing 263 values
- embarked: Missing 2 values

We'll remove any row that has even ONE missing value using `.dropna()`

````python
# STEP 4C: Remove rows with ANY missing values
# dropna() removes rows that contain NaN (missing values)
# We do this because our model can't handle missing data

rowsBeforeDrop = len(titanicData)  # Count rows before dropping
titanicData = titanicData.dropna()  # Remove rows with missing values
rowsAfterDrop = len(titanicData)  # Count rows after dropping

print(f"✓ Dropped rows with missing values")
print(f"Rows before: {rowsBeforeDrop}")
print(f"Rows after: {rowsAfterDrop}")
print(f"Rows removed: {rowsBeforeDrop - rowsAfterDrop}")
print(f"\nDataset shape after cleaning: {titanicData.shape}")
````

## Step 5: Encode Categorical Data

Machine learning models only understand numbers. We have two text columns:
- `sex`: "male" or "female"
- `embarked`: "C", "Q", or "S"

We'll convert these to numbers using **one-hot encoding**.

````python
# STEP 5: One-hot encode categorical columns
# pd.get_dummies() converts text categories into numeric columns (0 or 1)
# drop_first=True removes one column to avoid multicollinearity

print("Before encoding:")
print(titanicData[['sex', 'embarked']].head())  # Show original text values

# Encode the categorical columns
titanicData = pd.get_dummies(titanicData, columns=['sex', 'embarked'], drop_first=True)

print("\nAfter encoding:")
print(titanicData.head())  # Show encoded numeric values
print(f"\nNew columns created: {list(titanicData.columns)}")
````

### How One-Hot Encoding Works

**Before**:
```
sex        embarked
male       C
female     S
male       Q
```

**After** (with drop_first=True):
```
sex_male   embarked_Q   embarked_S
1          0            0
0          0            1
1          0            0
```

**Key points**:
- Each category becomes a binary column (1 = yes, 0 = no)
- `drop_first=True` removes redundancy (if sex_male=0, then it's female)
- Now all data is numeric!

## Step 6: Separate Features (X) and Target (y)

Machine learning models need two things:
- **X (features)**: All columns EXCEPT the target
- **y (target)**: The column we're trying to predict

For the Titanic data:
- **X**: pclass, age, sibsp, parch, fare, sex_male, embarked_Q, embarked_S
- **y**: survived

````python
# STEP 6: Separate features (X) and target (y)
# X = all features (everything except 'survived')
# y = target (what we're predicting)

X = titanicData.drop('survived', axis=1)  # Drop 'survived' column to get features
y = titanicData['survived']               # Extract 'survived' column as target

print("✓ Separated features and target")
print(f"Features (X) shape: {X.shape}")
print(f"Features (X) columns: {list(X.columns)}")
print(f"\nTarget (y) shape: {y.shape}")
print(f"Target (y) values: {y.unique()}")
print(f"\nTarget distribution:")
print(y.value_counts())  # Count how many survived vs didn't
````

### Why separate X and y?

During training, we show the model:
- **X**: "Here are the passenger features"
- **y**: "And here's whether they survived"

The model learns the relationship: features → survival

During prediction, we only have new X data and use the trained model to guess y!

## Step 7: Split Data into Training and Testing Sets

We divide our data into two parts:
- **Training set (80%)**: Used to teach the model
- **Testing set (20%)**: Used to evaluate the model

Why? We need to test on data the model has never seen before!

If we tested on training data, the model would memorize answers (overfitting).

````python
# STEP 7: Split data into training and testing sets
# test_size=0.2 means 20% testing, 80% training
# random_state=42 ensures reproducible results (same split every time)

xTrain, xTest, yTrain, yTest = train_test_split(
    X, y,                    # Features and target
    test_size=0.2,           # 20% for testing, 80% for training
    random_state=42          # For reproducibility
)

print("✓ Data split into train/test sets")
print(f"Training set size: {xTrain.shape[0]} rows ({len(xTrain)/len(X)*100:.1f}%)")
print(f"Testing set size: {xTest.shape[0]} rows ({len(xTest)/len(X)*100:.1f}%)")
print(f"\nTotal rows used: {len(xTrain) + len(xTest)}")
````

### Train vs Test Split

```
All Data (628 rows)
    |
    +---> Training Set (502 rows, 80%) 
    |     Used to teach the model
    |     Model learns patterns from this
    |
    +---> Testing Set (126 rows, 20%)
          Used to evaluate the model
          We check if model generalizes to new data
```

Why 80/20? Industry standard. Could be 70/30 or 75/25 too.

## Step 8: Build and Train the Decision Tree Model

Now we build our machine learning model using DecisionTreeClassifier.

````python
# STEP 8: Create and train the decision tree model
# max_depth=4 limits the tree to 4 decision levels (prevents overfitting)
# random_state=42 ensures reproducibility

model = DecisionTreeClassifier(
    max_depth=4,             # Limit tree depth (fewer decisions = simpler model)
    random_state=42          # For reproducibility
)

# Train the model on the training data
model.fit(xTrain, yTrain)  # Model learns from xTrain and yTrain

print("✓ Decision tree model trained successfully!")
print(f"Model type: {type(model).__name__}")
print(f"Max depth: {model.max_depth}")
print(f"Number of features: {model.n_features_in_}")
print(f"Feature names: {X.columns.tolist()}")
````

### What does .fit() do?

**Before .fit()**: Empty model with no knowledge
```
model = DecisionTreeClassifier(max_depth=4)
(no decisions learned yet)
```

**After .fit()**: Model has learned decision rules
```
model.fit(xTrain, yTrain)
(model learns: "If fare > 30, ask about sex...")
```

The model examines xTrain and yTrain thousands of times to find the best questions!

## Step 9: Make Predictions

Now we use our trained model to make predictions on the test set.

````python
# STEP 9: Make predictions on the test set
# predict() returns 0 (did not survive) or 1 (survived) for each row

predictions = model.predict(xTest)  # Predict survival for each test passenger

print("✓ Predictions made!")
print(f"Number of predictions: {len(predictions)}")
print(f"Unique prediction values: {set(predictions)}")
print(f"\nFirst 20 predictions: {predictions[:20]}")
print(f"\nHow many predicted to survive? {sum(predictions)}")
print(f"How many predicted to NOT survive? {len(predictions) - sum(predictions)}")
````

### What happens during prediction?

For each passenger in xTest:
1. Model asks: "Is fare > $30?" (Look at the tree)
2. If yes, asks: "Is sex female?" (Go left branch)
3. Keeps asking questions until reaching a leaf node
4. Leaf node says: "Predict: Survived" or "Predict: Did not survive"

All this happens automatically with .predict()!

## Step 10: Evaluate Model Performance

We need to know: How good is our model?

We'll use two metrics:
1. **Accuracy**: Percentage of correct predictions
2. **Classification Report**: Detailed breakdown (precision, recall, f1-score)

````python
# STEP 10A: Calculate accuracy
# accuracy_score() compares predictions to actual values
# Returns: (number correct) / (total predictions)

accuracy = accuracy_score(yTest, predictions)  # Compare predictions to actual values

print("="*80)
print("MODEL EVALUATION")
print("="*80)
print(f"Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"\nInterpretation:")
print(f"  - Out of {len(yTest)} test passengers")
print(f"  - Model correctly predicted {int(accuracy*len(yTest))} survivors")
print(f"  - Model made {len(yTest) - int(accuracy*len(yTest))} mistakes")
````

### Instructor Note

Accuracy tells us the overall correctness. For Titanic:
- Random guessing (50/50): ~50% accuracy
- Our model: ~80-85% accuracy

That's much better! But is it good enough? Depends on the use case.

````python
# STEP 10B: Print detailed classification report
# This shows precision, recall, and f1-score for each class

print("\n" + "="*80)
print("CLASSIFICATION REPORT")
print("="*80)
print(classification_report(yTest, predictions, target_names=['Did Not Survive', 'Survived']))
````

### Understanding the Classification Report

**Precision**: Of all passengers we predicted would survive, how many actually did?
- High precision = Few false positives

**Recall**: Of all passengers who actually survived, how many did we find?
- High recall = Few false negatives

**F1-Score**: Balance between precision and recall
- Harmonic mean of precision and recall

**Support**: How many examples in each class

## Step 11: Visualize the Decision Tree

Let's draw a picture of the actual decision tree the model created!

````python
# STEP 11: Visualize the decision tree
# plot_tree() creates a visual representation of the tree

plt.figure(figsize=(20, 10))  # Create large figure for tree
plot_tree(
    model,                                           # The trained model
    feature_names=X.columns,                         # Names of the features
    class_names=['Did Not Survive', 'Survived'],     # Names of the output classes
    filled=True,                                     # Color the nodes by class
    rounded=True,                                    # Round corners on boxes
    fontsize=10                                      # Font size for text
)
plt.title("Decision Tree for Titanic Survival Prediction\n(max_depth=4)", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

print("✓ Tree visualization complete!")
````

### Reading the Tree

Each box in the tree shows:
1. **Decision question** (e.g., "fare <= 19.3?")
2. **gini**: Impurity measure (lower = more pure/certain)
3. **samples**: How many passengers in this branch
4. **value**: [did_not_survive, survived] count
5. **Color**: Blue = mostly survived, Orange = mostly didn't

**Following a path**:
- Start at the top
- At each node, follow the YES (left) or NO (right) branch
- Continue until you reach a leaf (end) node
- Leaf shows the prediction

Not all features are equally important! Let's see which features the model uses most.

````python
# STEP 12: Calculate feature importance
# The model calculates how much each feature helps with decisions
# Higher importance = more useful for prediction

importances = model.feature_importances_  # Get importance scores

# Create a list of (feature_name, importance) pairs
featureImportance = list(zip(X.columns, importances))
featureImportance.sort(key=lambda x: x[1], reverse=True)  # Sort by importance (highest first)

print("="*80)
print("FEATURE IMPORTANCE")
print("="*80)
for feature, importance in featureImportance:
    # Display feature name, importance value, and a visual bar
    bar = '█' * int(importance * 50)  # Create a bar chart
    print(f"{feature:15} | {importance:6.4f} | {bar}")

print(f"\nTotal importance: {sum(importances):.4f}")
````

## Step 13: Make Predictions on New Passengers

Now let's predict survival for passengers not in our test set!

````python
# STEP 13: Make predictions on new passengers
# Create new passenger data (must have same format as training data!)

# New passenger 1: 1st class female, age 25, fare $100, 1 sibling, 0 parents, boarded from C
newPassenger1 = pd.DataFrame({
    'pclass': [1],
    'age': [25],
    'sibsp': [1],
    'parch': [0],
    'fare': [100],
    'sex_male': [0],      # female (1=yes, 0=no)
    'embarked_Q': [0],
    'embarked_S': [0]     # Boarded from C (neither Q nor S)
})

# New passenger 2: 3rd class male, age 30, fare $7, 0 siblings, 0 parents, boarded from S
newPassenger2 = pd.DataFrame({
    'pclass': [3],
    'age': [30],
    'sibsp': [0],
    'parch': [0],
    'fare': [7],
    'sex_male': [1],      # male
    'embarked_Q': [0],
    'embarked_S': [1]     # Boarded from S
})

# Make predictions
pred1 = model.predict(newPassenger1)[0]
pred2 = model.predict(newPassenger2)[0]

# Get prediction probabilities (confidence)
prob1 = model.predict_proba(newPassenger1)[0]
prob2 = model.predict_proba(newPassenger2)[0]

print("="*80)
print("PREDICTIONS FOR NEW PASSENGERS")
print("="*80)
print("\nPassenger 1: 1st class female, age 25, $100 fare")
print(f"  Prediction: {'SURVIVED ✓' if pred1 == 1 else 'DID NOT SURVIVE ✗'}")
print(f"  Confidence: {prob1[pred1]*100:.2f}%")

print("\nPassenger 2: 3rd class male, age 30, $7 fare")
print(f"  Prediction: {'SURVIVED ✓' if pred2 == 1 else 'DID NOT SURVIVE ✗'}")
print(f"  Confidence: {prob2[pred2]*100:.2f}%")
````

### Understanding Probability Predictions

**predict_proba()** gives us confidence scores:
- Returns [prob_0, prob_1] where:
  - prob_0 = probability of not surviving (class 0)
  - prob_1 = probability of surviving (class 1)

**Example**:
- [0.15, 0.85] = 15% chance died, 85% chance survived
- [0.90, 0.10] = 90% chance died, 10% chance survived

## Step 14: Compare Different Hyperparameters

Let's see how max_depth affects model performance!

````python
# STEP 14: Test different max_depth values
# We'll build multiple models and compare their accuracy

max_depths = [2, 3, 4, 5, 6, 7, 8]
results = []  # Store results for comparison

for depth in max_depths:
    # Build model with this max_depth
    tempModel = DecisionTreeClassifier(max_depth=depth, random_state=42)
    tempModel.fit(xTrain, yTrain)  # Train
    
    # Get predictions and accuracy
    trainPred = tempModel.predict(xTrain)
    testPred = tempModel.predict(xTest)
    
    trainAcc = accuracy_score(yTrain, trainPred)
    testAcc = accuracy_score(yTest, testPred)
    
    results.append({
        'max_depth': depth,
        'train_accuracy': trainAcc,
        'test_accuracy': testAcc
    })
    
    print(f"max_depth={depth}: Train={trainAcc:.4f}, Test={testAcc:.4f}")

print("\n" + "="*80)
print("INTERPRETATION")
print("="*80)
print("Notice:")
print("- As max_depth increases, training accuracy increases")
print("- Test accuracy increases then decreases (overfitting!)")
print("- Best max_depth balances train and test accuracy")
````

### Overfitting Explained

```
Accuracy
  ▲
  │     Training accuracy
  │      /
  │    /
  │  / Test accuracy (overfitting happens here!)
  │/
  └─────────────────→ max_depth
       Best point
```

**The problem**: 
- Shallow tree (depth=2): Underfits (too simple)
- Deep tree (depth=10): Overfits (memorizes training data)
- Sweet spot (depth=4-5): Best generalization
