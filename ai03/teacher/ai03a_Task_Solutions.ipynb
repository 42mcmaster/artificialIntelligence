```python
import pandas as pd
import matplotlib.pyplot as plt

print("✓ Libraries imported!")
```

```python
titanicData = pd.read_csv('Titanic Dataset.csv')
print(f"Dataset loaded: {titanicData.shape[0]} rows, {titanicData.shape[1]} columns")
```

```python
print("First 5 rows:")
print(titanicData.head())
```

```python
print("\nLast 5 rows:")
print(titanicData.tail())
```

```python
print("\nDataset Information:")
print(titanicData.info())
```

```python
print("\nStatistical Summary:")
print(titanicData.describe())
```

```python
totalPassengers = len(titanicData)
print(f"Q1: Total passengers = {totalPassengers}")
```

```python
totalColumns = len(titanicData.columns)
print(f"Q2: Total columns = {totalColumns}")
```

```python
avgAge = titanicData['age'].mean()
print(f"Q3: Average age = {avgAge:.2f}")
```

```python
survivorCount = titanicData['survived'].sum()
print(f"Q4: Survivors = {int(survivorCount)}")
```

```python
survivalRate = (titanicData['survived'].sum() / len(titanicData)) * 100
print(f"Q5: Survival rate = {survivalRate:.2f}%")
```

```python
print("Missing values per column:")
print(titanicData.isnull().sum())
print(f"\nTotal missing values: {titanicData.isnull().sum().sum()}")
```

```python
columnsNeeded = ['pclass', 'survived', 'sex', 'age', 'sibsp', 'parch', 'fare', 'embarked']
titanicData = titanicData[columnsNeeded]

print(f"✓ Selected {len(columnsNeeded)} columns")
print(f"Columns: {list(titanicData.columns)}")
```

```python
rowsBefore = len(titanicData)
titanicData = titanicData.dropna()
rowsAfter = len(titanicData)

print(f"✓ Dropped missing values")
print(f"Rows before: {rowsBefore}")
print(f"Rows after: {rowsAfter}")
print(f"Rows removed: {rowsBefore - rowsAfter}")
```

```python
print(f"\nMissing values after cleaning: {titanicData.isnull().sum().sum()}")
print("✓ Data is clean!")
```

```python
print("Before encoding:")
print(titanicData[['sex', 'embarked']].head())

titanicData = pd.get_dummies(titanicData, columns=['sex', 'embarked'], drop_first=True)

print("\nAfter encoding:")
print(titanicData.head())
print(f"\nNew shape: {titanicData.shape}")
print(f"New columns: {list(titanicData.columns)}")
```

```python
print("Data types:")
print(titanicData.dtypes)

print("\nData sample:")
print(titanicData.head())

print(f"\nFinal dataset shape: {titanicData.shape}")
print("✓ Data is ready for machine learning!")
```
