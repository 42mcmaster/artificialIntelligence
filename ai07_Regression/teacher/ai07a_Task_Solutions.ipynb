```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded!")
```

## Part 1: Height vs Shoe Size

```python
heightShoe = pd.read_excel('ai06_excel_height_shoe.xlsx')
print("First 10 rows:")
print(heightShoe.head(10))
print(f"\nShape: {heightShoe.shape}")
```

```python
# Calculate Pearson r
heightShoe_columns = heightShoe.columns
print(f"Columns: {list(heightShoe_columns)}")

# Assuming columns are 'Height' and 'Shoe Size' or similar
# Adjust if needed
if 'Height' in heightShoe.columns and 'Shoe Size' in heightShoe.columns:
    r_height_shoe = heightShoe['Height'].corr(heightShoe['Shoe Size'])
elif len(heightShoe.columns) == 2:
    r_height_shoe = heightShoe.iloc[:, 0].corr(heightShoe.iloc[:, 1])

print(f"Pearson r (Height vs Shoe Size): {r_height_shoe:.4f}")
```

```python
# Calculate R²
r2_height_shoe = r_height_shoe ** 2
print(f"R² (Height vs Shoe Size): {r2_height_shoe:.4f}  ({r2_height_shoe*100:.1f}%)")
```

```python
# Scatter plot
plt.figure(figsize=(8, 6))
if 'Height' in heightShoe.columns and 'Shoe Size' in heightShoe.columns:
    plt.scatter(heightShoe['Height'], heightShoe['Shoe Size'], alpha=0.5, color='steelblue', s=40)
    plt.xlabel('Height', fontsize=11)
    plt.ylabel('Shoe Size', fontsize=11)
else:
    plt.scatter(heightShoe.iloc[:, 0], heightShoe.iloc[:, 1], alpha=0.5, color='steelblue', s=40)
    plt.xlabel(heightShoe.columns[0], fontsize=11)
    plt.ylabel(heightShoe.columns[1], fontsize=11)

plt.title('Height vs Shoe Size', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nPearson r: {r_height_shoe:.4f}")
print(f"R²: {r2_height_shoe:.4f}")
```

### Instructor Note
**Expected answer for Height vs Shoe Size:**
- r ≈ 0.99 (very close to perfect)
- R² ≈ 0.98 (98% of shoe size explained by height)
- Scatter plot should show an almost perfect line
- This is a positive correlation
- Very strong relationship

## Part 4: Ranking All Three

```python
# Create ranking table
rankingData = [
    ('Height vs Shoe Size', r_height_shoe, r2_height_shoe, 'Very Strong'),
    ('Ice Cream Sales vs Temperature', r_ice_temp, r2_ice_temp, 'Very Strong'),
    ('Study Hours vs Test Score', r_study_score, r2_study_score, 'Very Strong')
]

# Sort by Pearson r (descending)
rankingData_sorted = sorted(rankingData, key=lambda x: x[1], reverse=True)

print("\nRanking by Strength of Correlation:\n")
for i, (rel, r, r2, strength) in enumerate(rankingData_sorted, 1):
    print(f"{i}. {rel}")
    print(f"   r = {r:.4f}, R² = {r2:.4f} ({r2*100:.1f}%)")
    print()
```

1. Height vs Shoe Size: r ≈ 0.99

3. Study Hours vs Test Score: r ≈ 0.97

## Part 5: Challenge — Weather Data

```python
# Load weather data
weather = pd.read_csv('medina_weather_2024.csv')
weather = weather.dropna()

print(f"Weather data shape: {weather.shape}")
print(f"\nAvailable columns: {list(weather.columns)}")
print(f"\nBasic stats:")
print(weather.describe().round(1))
```

```python
# Solution example: Compare temp_max with humidity and wind_speed
# (Students could choose any two)

print("SOLUTION EXAMPLE: Comparing temp_max with humidity and wind_speed\n")

r_temp_humidity = weather['temp_max'].corr(weather['humidity'])
r_temp_wind = weather['temp_max'].corr(weather['wind_speed'])

print(f"Pearson r (temp_max vs humidity): {r_temp_humidity:.4f}")
print(f"Pearson r (temp_max vs wind_speed): {r_temp_wind:.4f}")

print(f"\nR² values:")
print(f"  temp_max & humidity: {r_temp_humidity**2:.4f} ({(r_temp_humidity**2)*100:.1f}%)")
print(f"  temp_max & wind_speed: {r_temp_wind**2:.4f} ({(r_temp_wind**2)*100:.1f}%)")

print(f"\nStronger relationship: temp_max & humidity" if abs(r_temp_humidity) > abs(r_temp_wind) else "\nStronger relationship: temp_max & wind_speed")
```

```python
# Visualize both relationships
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: temp_max vs humidity
ax1.scatter(weather['humidity'], weather['temp_max'], alpha=0.4, color='steelblue')
ax1.set_xlabel('Humidity (%)', fontsize=11)
ax1.set_ylabel('Max Temperature (°F)', fontsize=11)
ax1.set_title(f'Max Temp vs Humidity (r={r_temp_humidity:.3f})', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)

# Plot 2: temp_max vs wind_speed
ax2.scatter(weather['wind_speed'], weather['temp_max'], alpha=0.4, color='steelblue')
ax2.set_xlabel('Wind Speed (mph)', fontsize=11)
ax2.set_ylabel('Max Temperature (°F)', fontsize=11)
ax2.set_title(f'Max Temp vs Wind Speed (r={r_temp_wind:.3f})', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Instructor Note
**Expected results:**
- Humidity: r ≈ -0.24 (weak negative — more humidity = slightly cooler)
- Wind speed: r ≈ -0.17 (weak negative — more wind = slightly cooler)
- Both much weaker than the excel datasets

Key teaching point: Real weather data is messier than carefully curated excel examples. Multiple factors affect temperature, so no single variable explains it well. This is why we use multiple features in regression (as we did in the walkthrough).
