## Summary

**What you've learned in this task:**

1. ✅ How to distinguish between regression (predicting numbers) and classification (predicting categories)
2. ✅ Real-world examples of each approach
3. ✅ When to use each type of machine learning
4. ✅ When complex problems need neural networks

**Key Question to Always Ask:**

"Is the output a **CATEGORY** (yes/no, spam/not spam, A/B/C/D/F) or a **NUMBER** (price, temperature, count)?"

- Category? → Classification
- Number? → Regression
- Complex (images, video, audio, language)? → Neural Network
