## Task 1 Solutions: Identify Regression vs. Classification

### Quick Decision Rule

**CLASSIFICATION:** Output is a CATEGORY (yes/no, A/B/C, spam/not spam, apple/banana/orange)

**REGRESSION:** Output is a NUMBER (price, rating, temperature, count)

### Problem 1: Movie Rating Prediction

**Answer:** REGRESSION

**Explanation:** The output is a number (stars: 1.0, 2.5, 3.5, 5.0). When predicting numbers, use regression.

**Note:** Even though ratings are often 1-5, they're still numbers (continuous). The ML model learns the relationship between user features and their likely rating.

### Problem 2: Email Spam Detection

**Answer:** CLASSIFICATION

**Explanation:** The output is a category (spam or not spam = 2 classes). Binary classification is when you predict one of two categories.

### Problem 3: House Price Prediction

**Answer:** REGRESSION

**Explanation:** The output is a number (dollar amount). Price can be any value, so it's regression.

### Problem 4: Student Grade Classification

**Answer:** CLASSIFICATION

**Explanation:** The output is a category (A, B, C, D, or F = 5 classes). Multi-class classification is when you predict one of many categories.

**Note:** Don't confuse this with numerical grades (like 87.5). This problem asks for LETTER grades, which are categories.

### Problem 5: Customer Churn Prediction

**Answer:** CLASSIFICATION

**Explanation:** The output is a category (will cancel or will stay = 2 classes). Binary classification.

### Problem 6: Salary Prediction

**Answer:** REGRESSION

**Explanation:** The output is a number (dollar amount). Salary predictions are continuous values, so use regression.

### Problem 7: Image Classification (Medical)

**Answer:** CLASSIFICATION (or Neural Network)

**Explanation:** The output is a category (broken bone or no break = 2 classes). This IS classification. However, since it involves images, a neural network (deep learning) would typically be used for the actual implementation. Accept either "classification" or "neural network."

### Problem 8: Stock Price Prediction

**Answer:** REGRESSION

**Explanation:** The output is a number (dollar amount). Stock prices are continuous values.

## Task 2 Solutions: Real-World AI to Approaches

### Real-World AI #1: Spotify Playlist Recommendation

**Answer:** Classification (can also say Neural Network)

**Explanation:** For each song, Spotify predicts: "Will this user like this song?" (Yes/No = binary classification). Modern Spotify uses deep learning, but the core task is classification. Accept either answer.

**Why NOT Regression?** The output isn't a rating number; it's a recommendation (yes/no).

### Real-World AI #2: Face Recognition on Facebook

**Answer:** Neural Network (or Deep Learning / Classification)

**Explanation:** Face recognition involves complex image analysis. This requires neural networks because:
- It's image processing (complex visual patterns)
- Simple classification/regression can't handle face variations (angles, lighting, expressions)
- Neural networks with multiple layers can learn subtle facial features

**Technically:** It's still classification (which person is this?), but neural networks do the classification.

### Real-World AI #3: Amazon Price Prediction

**Answer:** Regression

**Explanation:** The output is a number (price: $29.99, $34.99, $49.99). Even though prices are discrete dollar amounts, it's still regression because the model predicts a continuous numerical value.

### Real-World AI #4: Google Translate

**Answer:** Neural Network (Deep Learning)

**Explanation:** Language translation is extremely complex and requires neural networks:
- Must understand grammar, context, idioms
- Many-to-many sequence mapping (words → words)
- Simple classification/regression can't capture language complexity
- Modern translation uses transformer neural networks

**Not classification or basic regression** because translation generates new text, not just a category or number.

### Real-World AI #5: Credit Score Decision

**Answer:** Classification

**Explanation:** The output is a category (approve or deny = 2 classes). Binary classification.

## Common Student Errors

### Error 1: "Predicting a rating (1-5 stars) is classification because it's limited"
**Correction:** No, it's regression. Even though ratings are typically 1-5, the output is a CONTINUOUS number. Regression doesn't mean unlimited range—it means predicting a numerical value.

### Error 2: "If it has two categories (yes/no), it's always classification"
**Correction:** Correct! But some students confuse yes/no questions. If output is yes/no = classification. If output is a number = regression. Always look at WHAT is being predicted.

### Error 3: "Image recognition is always neural networks, never classification"
**Correction:** Image recognition IS classification (classifying images into categories). But neural networks are the best tool FOR that classification because images are complex.

### Error 4: "Medical diagnosis (disease or no disease) is regression"
**Correction:** No, it's classification. The output is a category (have disease / don't have disease), not a number.
