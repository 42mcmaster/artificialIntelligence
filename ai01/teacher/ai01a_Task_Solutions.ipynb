## Task 1 Solutions: Identify AI vs. Non-AI

### Scenario 1: YouTube Recommendations

**Correct Answer:** AI (Machine Learning)

**Explanation:** YouTube learns patterns from data about what users watch, not following hard-coded rules. The system improves as more viewing data is collected. It adapts to new users and new content without code changes. This is classic machine learning (collaborative filtering).

### Scenario 2: Age Verification on Websites

**Correct Answer:** Traditional Programming (NOT AI)

**Explanation:** This is a simple if/else rule that never changes. There's no learning from data. The programmer wrote the rule, and the computer follows it exactly. It's traditional programming.

```python
# Scenario 2 Example Code
def ageVerification(age):
    # Hard-coded rule: programmer writes the logic
    if age >= 18:
        return "Access granted"
    else:
        return "Access denied"

print(ageVerification(25))  # Access granted
print(ageVerification(16))  # Access denied
```

**Explanation:** The ATM uses hard-coded rules that the programmer wrote. It doesn't learn from new checks. It follows the same logic every time. Even though it's doing something complex (reading checks), it's not AI—it's just traditional programming with specific rules for digit recognition.

## Task 2 Solutions: Temperature-Based Clothing Recommendation

### Instructor Note

Students should write a function with if/elif/else statements. Accept reasonable variations. Key points:
1. The function uses hard-coded rules (traditional programming)
2. Each temperature range should have a different recommendation
3. Code should be readable with comments

```python
# SOLUTION: Clothing Recommendation Function

def recommendClothing(temperatureFahrenheit):
    """
    Input: Temperature in Fahrenheit (number)
    Output: Clothing recommendation (string)
    
    This is traditional programming with hard-coded rules.
    The programmer writes all the rules—no learning happens!
    """
    
    # Rule 1: Very hot weather
    if temperatureFahrenheit >= 90:
        return "Very hot! Wear shorts, t-shirt, and apply sunscreen."
    
    # Rule 2: Warm weather
    elif temperatureFahrenheit >= 70:
        return "Warm! Wear shorts or light pants and a t-shirt."
    
    # Rule 3: Cool weather
    elif temperatureFahrenheit >= 50:
        return "Cool! Wear pants and bring a light jacket."
    
    # Rule 4: Cold weather
    else:
        return "Cold! Wear a heavy jacket, long pants, hat, and gloves."

# Test the function
print("Temperature Recommendations:")
print("="*50)

testTemps = [95, 75, 60, 30]
for temp in testTemps:
    print(f"\nTemp: {temp}°F")
    print(f"Recommendation: {recommendClothing(temp)}")
```

### Expected Student Variations

Students might have:
- Different temperature thresholds (e.g., 85°F instead of 90°F) ✅ Acceptable
- Different clothing suggestions ✅ Acceptable as long as they're reasonable
- More or fewer temperature ranges ✅ Acceptable
- Better formatting or comments ✅ Great!

All of these are fine! The key is understanding they're writing rules, not building AI.

## Task 3 Solutions: Reflection Questions

### Question 1: Is Your Function AI?

**Expected Answer:** No, this is NOT AI. It's traditional programming.

**Why?**
- The programmer (student) wrote all the rules explicitly
- The function doesn't learn from data
- The function doesn't improve or adapt
- Same rules apply every time

**Teaching Note:** This is an important distinction! Many students confuse "complex logic" with "AI." Help them understand: if a human wrote the rules, it's traditional programming.

### Question 2: What Would Make It AI?

**Expected Answers (any of these):**

1. **Collect Training Data:**
   - Gather data about what thousands of people actually wear at each temperature
   - Include other factors: humidity, activity (exercise vs. sitting), personal preference

2. **Train a Model:**
   - Use machine learning to find patterns in the data
   - Let the algorithm discover the relationship between temperature/humidity and clothing choice

3. **Let it Learn:**
   - As more people provide examples, the AI improves
   - The AI adapts to new patterns (e.g., fashion trends, climate change)

**Teaching Note:** This shows students understand the core difference: AI learns from data, not from programmer rules.

### Question 3: Limitations of Rule-Based System

**Expected Answers (any of these):**

1. **Temperature isn't everything:**
   - High temperature + high humidity feels different than high temperature + low humidity
   - Wind speed, sun exposure, and cloud cover matter

2. **Personal differences:**
   - Some people feel cold easily, others feel hot easily
   - Different activities require different clothing

3. **Hard to update:**
   - If weather patterns change (climate change), code must be rewritten
   - If fashion changes, code must be updated

4. **Doesn't learn:**
   - If 1,000 people disagree with your rules, tough luck!
   - Can't adapt to new situations

5. **Inflexible:**
   - 69°F gets "light jacket" but 70°F gets "t-shirt"—seems arbitrary
   - Real people's preferences are more nuanced

**Teaching Note:** Connect this back to the spam filter example. Hard-coded rules don't scale. This is why AI is better for complex, real-world problems.

## Common Student Errors & How to Address Them

### Error 1: "YouTube is NOT AI because there are no neural networks"
**Correction:** YouTube doesn't need neural networks to be AI. It uses machine learning (collaborative filtering) to learn patterns from viewing data. ANY learning from data = AI.

### Error 2: "My function is AI because it's complex"
**Correction:** Complexity ≠ AI. If a human wrote all the rules, it's traditional programming, no matter how complex.

### Error 3: "Facial recognition is traditional programming because it's on my phone"
**Correction:** Where it runs doesn't matter. What matters is HOW it works. Facial recognition LEARNS from data = AI.

### Error 4: "Rule-based systems are a type of AI"
**Correction:** Hard-coded rule-based systems are NOT AI. They're traditional programming. AI requires LEARNING from data.
